## Pjesa 1 — Insertion Sort

### 1.1 Komentoni kodin

Shto komentin `#` pas cdo rreshti, duke shpjeguar cfare ben ai rresht konkret.

```python
def insertion_sort(lst):
    for i in range(1, len(lst)):           # 
        v, j = lst[i], i - 1              # 
        while j >= 0 and lst[j] > v:      # 
            lst[j + 1] = lst[j]           # 
            j -= 1                         # 
        lst[j + 1] = v                     # 
    return lst                             # 

# Testo kodin ketu:
print(insertion_sort([5, 3, 8, 1, 2]))
```

## Pjesa 2 — Selection Sort

### 2.1 Komentoni kodin

```python
def selection_sort(lista):
    n = len(lista)                              # 
    for i in range(n - 1):                     # 
        min_idx = i                             # 
        for j in range(i + 1, n):              # 
            if lista[j] < lista[min_idx]:      # 
                min_idx = j                    # 
        lista[i], lista[min_idx] = lista[min_idx], lista[i]  # 
    return lista                               # 

# Testo kodin ketu:
print(selection_sort([5, 3, 8, 1, 2]))
```

## Pjesa 3 — Merge Sort

### 3.1 Komentoni kodin

```python
def merge_sort(lista):
    if len(lista) <= 1:                                    # 
        return lista                                       # 
    mid = len(lista) // 2                                  # 
    left  = merge_sort(lista[:mid])                        # 
    right = merge_sort(lista[mid:])                        # 
    res, i, j = [], 0, 0                                   # 
    while i < len(left) and j < len(right):                # 
        if left[i] <= right[j]:                            # 
            res.append(left[i]); i += 1                    # 
        else:                                              # 
            res.append(right[j]); j += 1                  # 
    res.extend(left[i:]); res.extend(right[j:])            # 
    return res                                             # 

# Testo kodin ketu:
print(merge_sort([5, 3, 8, 1, 2]))
```
