## Algoritmet që trajtohen

Në këtë ushtrim do të shqyrtohen dhe do të implementohen disa nga algoritmet themelore të analizës së grafeve, të cilët shërbejnë si për eksplorimin ashtu edhe për optimizimin.

| Algoritmi | Funksioni kryesor |
|-----------|------------------|
| **BFS (Breadth-First Search)** | Gjen distancën më të shkurtër nga një nyje burim në grafet pa peshë |
| **DFS (Depth-First Search)** | Eksploron grafin, verifikon arritshmërinë dhe zbulon ciklet |
| **Kruskal / Prim** | Ndërtojnë pemën përfshirëse minimum në grafet me peshe |

## Ushtrimi 1 — Zgjidhja e labirintit me BFS dhe DFS

### Teoria: përshkimi i grafeve

Një **labirint** përbën një shembull tipik të një grafi të nënkuptuar: çdo qelizë e hapur shërben si kulm, ndërsa dy qeliza konsiderohen të lidhura me një brinjë nëse është e mundur të kalohet drejtpërdrejt nga njëra te tjetra. Algoritmet BFS dhe DFS janë dy mënyrat kryesore për të përshkuar një graf dhe për të gjetur një rrugë nga një kulm i caktuar në një tjetër.

---

### BFS (Breadth-First Search)

**Ideja kryesore:** shqyrtohen fillimisht të gjithë fqinjët e kulmit aktual, përpara se të kalohet në nivelin e ardhshëm të thellësisë.  
Mund të përfytyrohet si përhapja e një valë uji — ajo zgjerohet njëkohësisht në të gjitha drejtimet, rreth pas rrethi.

**Struktura e të dhënave:** radha — Queue (FIFO: i pari që hyn, i pari që del)

**Hapat e algoritmit:**
1. Vendoset kulmi fillestar në radhë dhe shënohet si i vizituar.
2. Nxirret kulmi nga fillimi i radhës dhe përpunohet.
3. Shtohen në radhë të gjithë fqinjët e pavizituar dhe shënohen si të tillë.
4. Veprimi përsëritet derisa radha të zbrazet ose të arrihet kulmi i kërkuar.

**Veti kryesore:** BFS gjen gjithmonë **rrugën më të shkurtër** (me numrin më të vogël të brinjëve) në grafet pa peshë.

```
Kompleksiteti kohor:  O(V + E)   ku V = numri i kulmeve, E = numri i brinjëve
```

---

### DFS (Depth-First Search)

**Ideja kryesore:** ndiqet një rrugë sa më thellë të jetë e mundur, përpara se të kthehemi pas për të provuar një tjetër.  
Mund të përfytyrohet si eksplorimi i një shpelle — fillimisht ndiqet një tunel deri në fund dhe vetëm pas kësaj kalohet në një tjetër.

**Struktura e të dhënave:** stiva — Stack (LIFO: i fundit që hyn, i pari që del), ose rekursion.

**Hapat e algoritmit:**
1. Vendoset kulmi fillestar në stive dhe shënohet si i vizituar.
2. Nxirret kulmi nga fillimi i stives dhe përpunohet.
3. Shtohen në stive të gjithë fqinjët e pavizituar.
4. Veprimi përsëritet derisa stiva të zbrazet ose të arrihet kulmi i kërkuar.

**Veti kryesore:** DFS **nuk e garanton** rrugën më të shkurtër, por shfrytëzon më pak memorie dhe zbatohet më thjesht në mënyrë rekursive.

```
Kompleksiteti kohor:  O(V + E)
```

---

### BFS kundrejt DFS — kur përdoret secili?

| Karakteristika | BFS | DFS |
|----------------|-----|-----|
| A e garanton rrugën më të shkurtër? | Po | Jo |
| Shfrytëzimi i memories | Më i lartë (grafe të gjera) | Më i ulët (grafe të thella) |
| Fusha të zbatimit | Rruga më e shkurtër, përshkimi sipas niveleve | Zbulimi i cikleve, renditja topologjike |
| Zbatimi teknik | Radhë | Stive / rekursion |

---

### Zbatimet në jetën reale

**BFS gjen zbatim në:**
- **navigimin GPS** (Google Maps) — për të përcaktuar rrugën me numrin më të vogël të kthesave ose ndalesave;
- **rrjetet sociale** — për sugjerimin e "personave që ndoshta i njihni" (miqtë e miqve);
- **robotikë** — për planifikimin e rrugës mbi një hartë të tipit rrjetë;
- **Web Crawlers** — për zbulimin e faqeve nivel pas niveli;
- **inteligjencën artificiale në lojëra** — për përcaktimin e mundësive të arritshme në lojërat strategjike.

**DFS gjen zbatim në:**
- **kompilatorë** — për zbulimin e varësive ciklike midis moduleve të kodit;
- **bazat e të dhënave** — për identifikimin e cikleve në marrëdhëniet me çelësa të huaj;
- **zgjidhësit e enigmave** — zgjidhjen e enigmave logjike;
- **analizën e rrjeteve** — për të gjetur grupe kulmesh të lidhura fort midis tyre.

**Si inxhinier, këto algoritme do t'i shfrytëzoni për:**
- projektimin e një rrjeti dhe verifikimin nëse të gjithë kulmet janë të arritshme;
- ndërtimin e sistemeve të tipit build system, ku duhet të zgjidhen varësitë midis komponentëve;
- ndërtimin e sistemeve të gjetjes së rrugës në video-lojëra;
- zhvillimin e sistemeve të rekomandimit, të cilat përshkojnë grafet e njohurive për t'i sugjeruar përdoruesit përmbajtje të lidhura me interesat e tij.

## Ushtrimi 1 — Zgjidhja e Labirintit me BFS dhe DFS


````python
import tkinter as tk
import threading
import time
from collections import deque

MAZE_TEXT = """\
###                 #########
#   ###################   # #
# ####          #     # # # #
# ########## ######## # # # #
#                     # # # #
##################### # # # #
#   ##                # #   #
# # ## ### ## ######### # # #
# #    #   ##B#         # # #
# # ## ################ # ###
# # ##             #### # # #
# # ############## ## # # # #
# #             ##    # # # #
# #### ######## ####### # # #
######    #             #   #
A      ######################"""

GAP = 2
COLORS = {
    "wall":      "#808080",
    "space":     "#000000",
    "start":     "#ff4500",
    "goal":      "#32cd32",
    "explored":  "#1e6fb5",   
    "path":      "#ffff00",   
    "frontier":  "#ff8800",   
    "bg":        "#000000",
    "btn_bg":    "#a6fabd",
    "btn_fg":    "#01671e",
    "btn_hover": "#01671e",
    "btn_h_fg":  "#a6fabd",
    "hover":     "#334433",
}

DIRS = [(-1, 0), (1, 0), (0, -1), (0, 1)]


class MazeSolver:
    def __init__(self, root):
        self.root = root
        self.root.title("Maze Solver")
        self.root.configure(bg=COLORS["bg"])

        # FIX C: use a threading.Event instead of a bare bool.
        self.solving_event = threading.Event()

        # FIX G: signal worker threads to abort when the window is closed.
        self.cancelled = threading.Event()

        self.step_mode  = tk.BooleanVar(value=False)
        self.step_event = threading.Event()

        # fullscreen
        self.root.attributes("-fullscreen", True)
        self.root.bind("<Escape>", self._on_escape)
        self.root.bind("<space>",  self._on_space)
        self.root.protocol("WM_DELETE_WINDOW", self._on_escape)
        self.root.update_idletasks()
        scr_w = self.root.winfo_screenwidth()
        scr_h = self.root.winfo_screenheight()

        # parse maze
        rows = MAZE_TEXT.split("\n")
        self.height = len(rows)
        self.width  = max(len(r) for r in rows)
        self.walls  = []
        self.open_cells = set()

        for i, row in enumerate(rows):
            wall_row = []
            for j in range(self.width):
                ch = row[j] if j < len(row) else " "
                if ch == "#":
                    wall_row.append(True)
                else:
                    wall_row.append(False)
                    self.open_cells.add((i, j))
            self.walls.append(wall_row)

        self.start   = None
        self.goal    = None
        self.placing = "start"   # "start" → "goal" → "ready"

        # compute cell size to fill screen
        title_space = 80
        btn_space   = 140
        avail_h = scr_h - title_space - btn_space
        avail_w = scr_w - 60

        cell_from_w = (avail_w - GAP) // self.width  - GAP
        cell_from_h = (avail_h - GAP) // self.height - GAP
        self.cell   = max(8, min(cell_from_w, cell_from_h))

        grid_w = self.width  * (self.cell + GAP) + GAP
        grid_h = self.height * (self.cell + GAP) + GAP

        # title
        font_size = max(18, self.cell * 2 // 3)
        tk.Label(root, text="BFS and DFS maze solving simulation",
                 font=("Helvetica", font_size, "bold"),
                 bg=COLORS["bg"], fg="white").pack(pady=(24, 12))

        # canvas
        self.canvas = tk.Canvas(root, width=grid_w, height=grid_h,
                                bg=COLORS["bg"], highlightthickness=0)
        self.canvas.pack()
        self.canvas.bind("<Button-1>", self._on_click)
        self.canvas.bind("<Motion>",   self._on_hover)

        # controls row
        btn_font = max(16, self.cell // 2)
        ctrl = tk.Frame(root, bg=COLORS["bg"])
        ctrl.pack(pady=(18, 0))

        for algo in ("BFS", "DFS"):
            tk.Button(
                ctrl, text=algo,
                font=("Courier New", btn_font),
                bg=COLORS["btn_bg"], fg=COLORS["btn_fg"],
                activebackground=COLORS["btn_hover"],
                activeforeground=COLORS["btn_h_fg"],
                relief="solid", bd=1, width=12,
                command=lambda a=algo: self.start_solve(a),
            ).pack(side="left", padx=10)

        tk.Checkbutton(
            ctrl, text="Step-by-step (Space)",
            variable=self.step_mode,
            font=("Courier New", max(12, btn_font - 4)),
            bg=COLORS["bg"], fg="#a6fabd",
            selectcolor=COLORS["bg"],
            activebackground=COLORS["bg"],
            activeforeground="#a6fabd",
        ).pack(side="left", padx=(24, 0))

        tk.Button(
            ctrl, text="Reset",
            font=("Courier New", btn_font),
            bg="#ff6666", fg="white",
            activebackground="#cc3333", activeforeground="white",
            relief="solid", bd=1, width=10,
            command=self._reset,
        ).pack(side="left", padx=(24, 0))

        # status bar
        self.status = tk.Label(root, text="",
                               font=("Courier New", max(12, btn_font - 4)),
                               bg=COLORS["bg"], fg="#888888")
        self.status.pack(pady=(10, 0))

        # legend
        legend = tk.Frame(root, bg=COLORS["bg"])
        legend.pack(pady=(4, 0))
        for label, colour in [("Explored", COLORS["explored"]),
                               ("Frontier", COLORS["frontier"]),
                               ("Path",     COLORS["path"])]:
            tk.Label(legend, text="■ ", fg=colour,
                     font=("Helvetica", 12), bg=COLORS["bg"]).pack(side="left")
            tk.Label(legend, text=label + "  ", fg="#888888",
                     font=("Helvetica", 12), bg=COLORS["bg"]).pack(side="left")

        # hint
        tk.Label(root, text="Escape = exit",
                 font=("Helvetica", 10), bg=COLORS["bg"],
                 fg="#333333").pack(side="bottom", pady=(0, 10))

        # draw
        self.cell_ids   = {}
        self.hover_cell = None
        self._draw_maze()
        self._set_status("Click a black cell to place the START point")

    # window / keyboard events

    def _on_escape(self, event=None):
        self.cancelled.set()
        self.step_event.set()
        self.root.destroy()

    def _on_space(self, event):
        self.step_event.set()

    # pixel → grid

    def _pixel_to_cell(self, px, py):
        j = (px - GAP) // (self.cell + GAP)
        i = (py - GAP) // (self.cell + GAP)
        if 0 <= i < self.height and 0 <= j < self.width:
            return i, j
        return None

    # hover highlight

    def _on_hover(self, event):
        if self.placing == "ready" or self.solving_event.is_set():
            return
        cell = self._pixel_to_cell(event.x, event.y)
        if cell == self.hover_cell:
            return
        if self.hover_cell and self.hover_cell in self.cell_ids:
            old = self.hover_cell
            if old != self.start and old != self.goal and old in self.open_cells:
                self.canvas.itemconfig(self.cell_ids[old], fill=COLORS["space"])
        self.hover_cell = cell
        if cell and cell in self.open_cells and cell != self.start and cell != self.goal:
            self.canvas.itemconfig(self.cell_ids[cell], fill=COLORS["hover"])

    # click to place start / goal 

    def _on_click(self, event):
        if self.solving_event.is_set():
            return
        cell = self._pixel_to_cell(event.x, event.y)
        if not cell or cell not in self.open_cells:
            return

        if self.placing == "start":
            if self.start and self.start in self.cell_ids:
                self.canvas.itemconfig(self.cell_ids[self.start], fill=COLORS["space"])
            self.start = cell
            self.canvas.itemconfig(self.cell_ids[cell], fill=COLORS["start"])
            self.placing = "goal"
            self._set_status("Now click a black cell to place the GOAL point")

        elif self.placing == "goal":
            if cell == self.start:
                return
            if self.goal and self.goal in self.cell_ids:
                self.canvas.itemconfig(self.cell_ids[self.goal], fill=COLORS["space"])
            self.goal = cell
            self.canvas.itemconfig(self.cell_ids[cell], fill=COLORS["goal"])
            self.placing = "ready"
            self._set_status("Ready — click BFS or DFS to solve!")

    # reset───

    def _reset(self):
        if self.solving_event.is_set():
            return
        self.start      = None
        self.goal       = None
        self.placing    = "start"
        self.hover_cell = None
        self._draw_maze()
        self._set_status("Click a black cell to place the START point")

    # drawing helpers

    def _cell_rect(self, i, j):
        x1 = GAP + j * (self.cell + GAP)
        y1 = GAP + i * (self.cell + GAP)
        return x1, y1, x1 + self.cell, y1 + self.cell

    def _draw_maze(self):
        self.canvas.delete("all")
        self.cell_ids = {}
        for i in range(self.height):
            for j in range(self.width):
                x1, y1, x2, y2 = self._cell_rect(i, j)
                if (i, j) == self.start:
                    c = COLORS["start"]
                elif (i, j) == self.goal:
                    c = COLORS["goal"]
                elif self.walls[i][j]:
                    c = COLORS["wall"]
                else:
                    c = COLORS["space"]
                rid = self.canvas.create_rectangle(
                    x1, y1, x2, y2, fill=c, outline="", width=0)
                self.cell_ids[(i, j)] = rid

    def _color_cell(self, i, j, color):
        self.canvas.itemconfig(self.cell_ids[(i, j)], fill=color)

    def _set_status(self, text):
        self.status.config(text=text)

    # wait helper──

    def _wait_step(self, explored_n, frontier_n):
        if self.cancelled.is_set():
            return
        if self.step_mode.get():
            self.root.after(0, self._set_status,
                f"Explored: {explored_n}  |  Frontier: {frontier_n}  |  Press Space →")
            self.step_event.clear()
            self.step_event.wait()
        else:
            self.root.after(0, self._set_status,
                f"Explored: {explored_n}  |  Frontier: {frontier_n}")
            time.sleep(0.04)

    # path reconstruction from parent map──────────────────────

    @staticmethod
    def _reconstruct(parent, goal):
        path, cur = [], goal
        while cur is not None:
            path.append(cur)
            cur = parent[cur]
        path.reverse()
        return path

    # FIX E: batch fade-out ──────
    # Collecting all cells to clear and issuing a *single* root.after call avoids
    # flooding the tkinter event queue with thousands of micro-callbacks, which
    # caused visible jank on large mazes in the original implementation.

    def _fade_explored(self, explored, path_set):
        cells_to_clear = [
            c for c in explored
            if c not in path_set and c != self.start and c != self.goal
        ]
        self.root.after(0, lambda: [
            self._color_cell(c[0], c[1], COLORS["space"])
            for c in cells_to_clear
        ])

    # entry point──

    def start_solve(self, algorithm):
        if self.solving_event.is_set():
            return
        if not self.start or not self.goal:
            self._set_status("Place both START and GOAL first!")
            return
        self.solving_event.set()
        self.step_event.clear()
        self._draw_maze()
        # FIX F: dispatch to dedicated BFS / DFS helpers rather than branching
        # inside a single merged _solve method.
        target = self._solve_bfs if algorithm == "BFS" else self._solve_dfs
        threading.Thread(target=target, daemon=True).start()

    # shared finish helper ───────

    def _finish(self, found_path, explored):
        if self.cancelled.is_set():
            self.solving_event.clear()
            return

        if found_path:
            self.root.after(0, self._set_status,
                f"Done!  Path length: {len(found_path)}  |  Explored: {len(explored)}")
            for cell in found_path[1:-1]:
                self.root.after(0, self._color_cell,
                                cell[0], cell[1], COLORS["path"])
        else:
            self.root.after(0, self._set_status,
                f"No path found!  Explored: {len(explored)}")

        time.sleep(0.8)

        # FIX E: single batched after() call instead of one per cell.
        path_set = set(found_path)
        self._fade_explored(explored, path_set)

        self.solving_event.clear()
        self.placing = "ready"

    def _solve_bfs(self):
        parent         = {self.start: None}
        queue          = deque([self.start])
        frontier_cells = {self.start}
        explored       = set()
        found_path     = []

        while queue and not self.cancelled.is_set():
            state = queue.popleft()           # FIFO — correct for BFS
            frontier_cells.discard(state)

            if state in explored:
                continue
            explored.add(state)

            if state != self.start and state != self.goal:
                self.root.after(0, self._color_cell,
                                state[0], state[1], COLORS["explored"])

            found = False
            for di, dj in DIRS:
                ni, nj    = state[0] + di, state[1] + dj
                neighbour = (ni, nj)

                if not (0 <= ni < self.height and 0 <= nj < self.width):
                    continue
                if self.walls[ni][nj]:
                    continue
                if neighbour in explored or neighbour in frontier_cells:
                    continue

                parent[neighbour] = state
                frontier_cells.add(neighbour)
                queue.append(neighbour)

                if neighbour == self.goal:
                    # Early exit: goal found at push time (optimal for BFS).
                    found_path = self._reconstruct(parent, self.goal)
                    found = True
                    break

                self.root.after(0, self._color_cell, ni, nj, COLORS["frontier"])

            if found:
                break
            self._wait_step(len(explored), len(queue))

        self._finish(found_path, explored)

    def _solve_dfs(self):
        parent       = {self.start: None}
        stack        = deque([self.start])
        dfs_on_stack = {self.start}
        explored     = set()
        found_path   = []

        while stack and not self.cancelled.is_set():
            state = stack.pop()               # LIFO — correct for DFS
            dfs_on_stack.discard(state)       # FIX A: release slot for backtrack

            if state in explored:
                continue
            explored.add(state)

            if state != self.start and state != self.goal:
                self.root.after(0, self._color_cell,
                                state[0], state[1], COLORS["explored"])

            found = False
            for di, dj in DIRS:
                ni, nj    = state[0] + di, state[1] + dj
                neighbour = (ni, nj)

                if not (0 <= ni < self.height and 0 <= nj < self.width):
                    continue
                if self.walls[ni][nj]:
                    continue
                if neighbour in explored:
                    continue
                if neighbour in dfs_on_stack:  # FIX A: skip if already stacked
                    continue

                # FIX D: always overwrite — keeps parent consistent with the
                # branch DFS is currently following, not an earlier abandoned one.
                parent[neighbour] = state
                dfs_on_stack.add(neighbour)
                stack.append(neighbour)

                if neighbour == self.goal:
                    found_path = self._reconstruct(parent, self.goal)
                    found = True
                    break

                self.root.after(0, self._color_cell, ni, nj, COLORS["frontier"])

            if found:
                break
            self._wait_step(len(explored), len(stack))

        self._finish(found_path, explored)


if __name__ == "__main__":
    root = tk.Tk()
    MazeSolver(root)
    root.mainloop()
````

## Ushtrimi 2 — Ndërtimi interaktiv i grafit me BFS dhe DFS

**Dallimi kryesor nga Ushtrimi 1:**

| Ushtrimi 1 — Labirinti | Ushtrimi 2 — Ndërtuesi interaktiv |
|------------------------|-----------------------------------|
| Graf i nënkuptuar (rrjetë dydimensionale) | Graf i paraqitur qartë (kulme dhe brinjë) |
| Pa peshë | I drejtuar (brinjët kanë drejtim) |
| Grafi i paracaktuar | Grafin e ndërtoni vetë |
| Vizualizim me ngjyra në rrjetë | Vizualizim me shigjeta midis kulmeve |
---

### Çfarë duhet vëzhguar

Ndërtoni të njëjtin graf dhe më pas ekzekutoni BFS dhe DFS duke u nisur nga i njëjti kulm fillestar. Vini re:

1. **Rendi i zbulimit** — BFS i viziton fillimisht kulmet më afër pikës së nisjes, ndërsa DFS zhytet menjëherë në thellësi.
2. **Rruga e gjetur** — BFS gjen gjithmonë rrugën më të shkurtër deri te destinacioni, ndërsa DFS mund të gjejë një rrugë më të gjatë.
3. **Numri i hapave** — Krahasoni sa hapa i nevojiten secilit algoritëm për të përfunduar.

## Ushtrimi 2 — Ndërtimi Interaktiv i Grafit me BFS dhe DFS


````python
import sys
import pygame
import math
from collections import deque

pygame.init()

info         = pygame.display.Info()
WIDTH, HEIGHT= info.current_w, info.current_h
screen       = pygame.display.set_mode((WIDTH, HEIGHT), pygame.FULLSCREEN)
pygame.display.set_caption("Interactive BFS / DFS Graph Builder")

FONT   = pygame.font.SysFont("Segoe UI", 16)

BG            = (15,  20,  35)
NODE_R        = 20
C_NODE        = (80,  100, 140)
C_NODE_VIS    = (40,  160, 80)
C_NODE_TARGET = (30,  120, 220)
C_EDGE        = (80,  90,  120)
C_EDGE_ACTIVE = (255, 80,  80)
C_PATH        = (255, 160, 30)
UI_BG         = (25,  30,  50)
BTN_BG        = (60,  80,  120)
BTN_HOVER     = (90,  120, 180)
BTN_TEXT      = (255, 255, 255)


def quit_game():
    pygame.quit()
    sys.exit()


# ─────────────────────────────────────────────
# ARROW DRAW
# ─────────────────────────────────────────────
def draw_arrow(surf, color, start, end, width=2, head=12, rad=0.13):
    dx, dy = end[0]-start[0], end[1]-start[1]
    if math.hypot(dx, dy) < 1:
        return
    mx = (start[0]+end[0])/2 + rad*(-dy)
    my = (start[1]+end[1])/2 + rad*(dx)
    pts = [
        (
            (1-t/20)**2*start[0] + 2*(1-t/20)*(t/20)*mx + (t/20)**2*end[0],
            (1-t/20)**2*start[1] + 2*(1-t/20)*(t/20)*my + (t/20)**2*end[1],
        )
        for t in range(21)
    ]
    if len(pts) > 1:
        pygame.draw.lines(surf, color, False,
                          [(int(x), int(y)) for x, y in pts], width)

    def head_draw(tx, ty, adx, ady):
        l = (tx - head*adx + head*0.4*(-ady),
             ty - head*ady + head*0.4*(adx))
        r = (tx - head*adx - head*0.4*(-ady),
             ty - head*ady - head*0.4*(adx))
        pygame.draw.polygon(surf, color, [(tx, ty), l, r])

    for i in range(len(pts)-1, 0, -1):
        px, py = pts[i]
        if math.hypot(px-end[0], py-end[1]) >= NODE_R+4:
            nx, ny   = pts[i-1]
            adx, ady = px-nx, py-ny
            l = math.hypot(adx, ady)
            if l < 0.1:
                break
            head_draw(px, py, adx/l, ady/l)
            break


# ─────────────────────────────────────────────
# BFS — FIFO via deque.popleft()
# ─────────────────────────────────────────────
def bfs(adj, s):
    n       = len(adj)
    marked  = [False] * n
    edge_to = [-1]    * n
    steps   = []

    q        = deque([s])
    marked[s]= True

    print("\n" + "="*60)
    print(f"BFS STARTED  —  source: {s}")
    print("="*60)
    for i, nb in enumerate(adj):
        print(f"  adj[{i}] = {nb}")
    print(f"  Initial queue (FIFO): {list(q)}")

    while q:
        v = q.popleft()
        steps.append(("visit", v, None, marked[:], edge_to[:]))
        print(f"\n  [VISIT] node {v}  |  queue: {list(q)}")

        for w in adj[v]:
            if not marked[w]:
                marked[w]  = True
                edge_to[w] = v
                q.append(w)
                steps.append(("edge", v, w, marked[:], edge_to[:]))
                print(f"    enqueue {v}→{w}  | queue: {list(q)}")
            else:
                steps.append(("skip", v, w, marked[:], edge_to[:]))
                print(f"    skip    {v}→{w}  (visited)")

    print("\n" + "="*60)
    print(f"BFS FINISHED  edge_to: {edge_to}")
    print("="*60 + "\n")
    return steps


# ─────────────────────────────────────────────
# DFS — LIFO via stack
# Push ALL neighbors at once.
# LIFO picks the LAST pushed neighbor first,
# which is the LAST edge you drew from that node.
# ─────────────────────────────────────────────
def dfs(adj, s):
    n       = len(adj)
    marked  = [False] * n
    edge_to = [-1]    * n
    steps   = []
    stack   = [s]

    print("\n" + "="*60)
    print(f"DFS STARTED  —  source: {s}")
    print("="*60)
    for i, nb in enumerate(adj):
        print(f"  adj[{i}] = {nb}")
    print(f"  Initial stack (LIFO): {stack}")

    while stack:
        v = stack.pop()                    # LIFO

        if marked[v]:
            print(f"\n  [POP] {v} already visited — skip")
            continue

        marked[v] = True
        steps.append(("visit", v, None, marked[:], edge_to[:]))
        print(f"\n  [VISIT] node {v}")
        print(f"    adj[{v}] = {adj[v]}")
        print(f"    pushing all unvisited neighbors → LIFO picks LAST one first")

        for w in adj[v]:
            if not marked[w]:
                edge_to[w] = v
                stack.append(w)
                steps.append(("edge", v, w, marked[:], edge_to[:]))
                print(f"    push {w}  | stack now: {stack}")

    print("\n" + "="*60)
    print(f"DFS FINISHED  edge_to: {edge_to}")
    print("="*60 + "\n")
    return steps


# ─────────────────────────────────────────────
# APP
# ─────────────────────────────────────────────
class App:
    def __init__(self):
        self.nodes   = []
        self.edges   = []     # LIST preserves draw order exactly
        self.history = []

        self.mode        = "build"
        self.edge_start  = None
        self.start       = None
        self.target      = None

        self.algo         = "BFS"
        self.steps        = []
        self.step_i       = 0
        self.control_mode = "play"
        self.play         = False
        self.speed        = 3
        self.timer        = 0

        self.btn_run   = pygame.Rect(WIDTH-380, HEIGHT-180, 300, 45)
        self.btn_reset = pygame.Rect(WIDTH-380, HEIGHT-120, 300, 45)

    def add_node(self, pos):
        nid = len(self.nodes)
        self.nodes.append(pos)
        self.history.append(("node", nid))
        print(f"\n[NODE ADDED]  id={nid}")

    def _edge_exists(self, u, v):
        return any(eu == u and ev == v for eu, ev in self.edges)

    def add_edge(self, u, v):
        if u == v:
            return
        if self._edge_exists(u, v):
            print(f"\n[EDGE SKIPPED]  {u}→{v}  duplicate")
            return
        self.edges.append((u, v))
        self.history.append(("edge", u, v))
        print(f"\n[EDGE ADDED]  {u}→{v}  order: {self.edges}")

    def undo(self):
        if not self.history:
            return
        last = self.history.pop()
        if last[0] == "node":
            idx = last[1]
            self.nodes.pop(idx)
            self.edges = [(u, v) for u, v in self.edges
                          if u != idx and v != idx]
        elif last[0] == "edge":
            self.edges.remove((last[1], last[2]))

    def clear(self):
        self.nodes.clear()
        self.edges.clear()
        self.history.clear()
        self.start  = None
        self.target = None
        self.reset_algo()
        print("\n[CLEAR]")

    def build_adj(self):
        adj = [[] for _ in self.nodes]
        for u, v in self.edges:
            adj[u].append(v)
        print("\n" + "-"*50)
        print("ADJACENCY LIST (matches draw order):")
        for i, nb in enumerate(adj):
            print(f"  adj[{i}] = {nb}")
        print("-"*50)
        return adj

    def reset_algo(self):
        self.step_i = 0
        self.play   = False
        self.timer  = 0
        self.steps  = []

    def run_algo(self):
        if self.start is None:
            return
        adj = self.build_adj()
        self.steps  = bfs(adj, self.start) if self.algo == "BFS" \
                      else dfs(adj, self.start)
        self.step_i = 0
        self.play   = False
        self.timer  = 0

    def next_step(self):
        if not self.steps:
            return
        if self.step_i < len(self.steps) - 1:
            self.step_i += 1
            self._print_step()
        else:
            print("\n[END OF STEPS]")

    def prev_step(self):
        if not self.steps:
            return
        if self.step_i > 0:
            self.step_i -= 1
            self._print_step()
        else:
            print("\n[ALREADY AT STEP 0]")

    def _print_step(self):
        kind, v, w, marked, edge_to = self.steps[self.step_i]
        visited = [i for i, m in enumerate(marked) if m]
        print(f"\n[STEP {self.step_i}/{len(self.steps)-1}]"
              f"  {kind:5s}  node={v}  neighbor={w}  visited={visited}")

    def get_path(self):
        if self.target is None or not self.steps:
            return []
        edge_to = self.steps[self.step_i][4]
        path, x = [], self.target
        seen = set()
        while x != -1:
            if x in seen:
                break
            seen.add(x)
            path.append(x)
            x = edge_to[x]
        path.reverse()
        if path and path[0] != self.start:
            return []
        return path

    def handle_click(self, pos):
        if self.btn_run.collidepoint(pos):
            self.run_algo()
            return
        if self.btn_reset.collidepoint(pos):
            self.reset_algo()
            return
        for i, (x, y) in enumerate(self.nodes):
            if math.hypot(pos[0]-x, pos[1]-y) < NODE_R:
                self.handle_node_click(i)
                return
        if self.mode == "build":
            self.add_node(pos)

    def handle_node_click(self, i):
        if self.mode == "build":
            if self.edge_start is None:
                self.edge_start = i
            else:
                self.add_edge(self.edge_start, i)
                self.edge_start = None
        elif self.mode == "start":
            self.start = i
            print(f"\n[START] node {i}")
            self.run_algo()
        elif self.mode == "target":
            self.target = i
            print(f"\n[TARGET] node {i}")

    def draw_button(self, rect, text):
        mouse = pygame.mouse.get_pos()
        color = BTN_HOVER if rect.collidepoint(mouse) else BTN_BG
        pygame.draw.rect(screen, color, rect, border_radius=8)
        pygame.draw.rect(screen, (120, 150, 220), rect, 2, border_radius=8)
        lbl = FONT.render(text, True, BTN_TEXT)
        screen.blit(lbl, lbl.get_rect(center=rect.center))

    def draw(self):
        screen.fill(BG)

        for u, v in self.edges:
            draw_arrow(screen, C_EDGE, self.nodes[u], self.nodes[v], 2)

        if self.steps:
            kind, v, w, marked, edge_to = self.steps[self.step_i]
            if w is not None:
                draw_arrow(screen, C_EDGE_ACTIVE,
                           self.nodes[v], self.nodes[w], 4)

        path = self.get_path()
        for i in range(len(path)-1):
            draw_arrow(screen, C_PATH,
                       self.nodes[path[i]], self.nodes[path[i+1]], 4)

        if self.edge_start is not None:
            mx, my = pygame.mouse.get_pos()
            draw_arrow(screen, (150, 150, 255),
                       self.nodes[self.edge_start], (mx, my), 2)

        for i, (x, y) in enumerate(self.nodes):
            col = C_NODE
            if self.steps and self.steps[self.step_i][3][i]:
                col = C_NODE_VIS
            if i == self.target:
                col = C_NODE_TARGET      # blue
            if i == self.start:
                col = (200, 60, 60)      # red — always on top
            pygame.draw.circle(screen, col, (x, y), NODE_R)
            lbl = FONT.render(str(i), True, (255, 255, 255))
            screen.blit(lbl, lbl.get_rect(center=(x, y)))

        pygame.draw.rect(screen, UI_BG, (WIDTH-400, 0, 400, HEIGHT))
        y = 10
        def t(s):
            nonlocal y
            screen.blit(FONT.render(s, True, (200, 220, 255)), (WIDTH-390, y))
            y += 22

        t(f"Mode      : {self.mode}")
        t(f"Algorithm : {self.algo}")
        t(f"Control   : {self.control_mode}")
        t(f"Step      : {self.step_i} / {max(0, len(self.steps)-1)}")
        t(f"Path      : {' → '.join(map(str, path)) if path else '—'}")
        t("")
        t("── Colors ────────────────────")
        t("Red    : start node")
        t("Blue   : target node")
        t("Green  : visited node")
        t("Orange : path found")
        t("")
        t("── Controls ──────────────────")
        t("B : build mode")
        t("S : set start node")
        t("T : set target node")
        t("D : use DFS")
        t("F : use BFS")
        t("M : toggle play / step")
        t("SPACE : play / pause")
        t("N : next step")
        t("P : prev step")
        t("Z : undo")
        t("C : clear graph")
        t("ESC : exit")

        self.draw_button(self.btn_run,   "Run Algorithm")
        self.draw_button(self.btn_reset, "Reset Steps")
        pygame.display.flip()

    def run(self):
        clock = pygame.time.Clock()
        print("\n" + "="*60)
        print("  BFS / DFS Visualizer  —  press ESC to exit")
        print("="*60)

        while True:
            dt = clock.tick(60) / 1000

            if self.control_mode == "play" and self.play and self.steps:
                self.timer += dt
                if self.timer > 1 / self.speed:
                    self.timer = 0
                    if self.step_i < len(self.steps) - 1:
                        self.step_i += 1
                    else:
                        self.play = False
                        print("\n[AUTO]  paused at last step")

            for e in pygame.event.get():

                if e.type == pygame.QUIT:
                    quit_game()

                if e.type == pygame.KEYDOWN:
                    if e.key == pygame.K_ESCAPE:
                        quit_game()                        # clean exit
                    elif e.key == pygame.K_m:
                        self.control_mode = (
                            "step" if self.control_mode == "play" else "play")
                    elif e.key == pygame.K_SPACE:
                        if self.control_mode == "play":
                            self.play = not self.play
                    elif e.key == pygame.K_n:
                        if self.control_mode == "step":
                            self.next_step()
                    elif e.key == pygame.K_p:
                        if self.control_mode == "step":
                            self.prev_step()
                    elif e.key == pygame.K_b:
                        self.mode = "build"
                    elif e.key == pygame.K_s:
                        self.mode = "start"
                    elif e.key == pygame.K_t:
                        self.mode = "target"
                    elif e.key == pygame.K_d:
                        self.algo = "DFS"
                        print("\n[ALGO]  DFS")
                    elif e.key == pygame.K_f:
                        self.algo = "BFS"
                        print("\n[ALGO]  BFS")
                    elif e.key == pygame.K_z:
                        self.undo()
                    elif e.key == pygame.K_c:
                        self.clear()

                if e.type == pygame.MOUSEBUTTONDOWN:
                    self.handle_click(e.pos)

            self.draw()


if __name__ == "__main__":
    App().run()
````

## Ushtrimi 3 — Pema Perfshirese Minimum: Kruskal dhe Prim

### Çfarë është një Pemë Perfshirese?

Duke pasur një **graf të lidhur, pa peshë, të pa drejtuar** me N kulme, një **Pemë Perfshirese** është nëngrafi që:
- Lidh **të gjitha kulmet** (mbulon grafin)
- Përmban saktësisht **N − 1 brinjë** (pa cikle)
- Është një **pemë** (e lidhur dhe acilike)

Një **Pemë Perfshirese Minimum (MST)** është pema perfshirese ku **shuma totale e peshëve të brinjëve është minimale**.

> - **Analogji:** Imagino se je planifikues qyteti dhe duhet të lidhësh 5 lagje me rrugë. Dëshiron që çdo lagje të jetë e arritshme nga çdo tjetër, por dëshiron të ndërtosh sa më pak kilometre rrugë. MST jep zgjidhjen optimale!

---

### Algoritmi i Kruskalit (Bazuar në brinjë / Greedy)

**Ideja kryesore:** Renditi të gjitha brinjët sipas peshës (greedy) shto brinjën me peshë më të vogël që nuk krijon cikël.

**Hapat e algoritmit:**
1. Renditi të gjitha brinjët sipas peshës në rend rritës.
2. Zgjidh brinjën më të vogël. Nëse lidh dy komponente të ndryshme (pa cikël) → shtohet në MST.
3. Përsërit derisa MST të ketë N − 1 brinjë.

**Struktura kryesore:** **Bashkim-Gjej (DSU / Union-Find)** — kontrollon me efikasitet nëse dy nyje janë në të njëjtin komponent.

**Kruskali funksionon më mirë** kur grafi është **i rrallë** (pak brinjë).

---

### Algoritmi i Prim (Bazuar në kulme / Greedy)

**Ideja kryesore:** Rrit MST-n një kulm në të njëjtën kohë. Gjithmonë zgjidhni brinjën me peshë më të vogël që lidh pemën aktuale me një kulm të ri.

**Hapat e algoritmit:**
1. Fillo nga çdo kulm. Shtohet në MST.
2. Shiko të gjitha brinjët që largohen nga MST aktual.
3. Zgjidh brinjën me peshën minimale që shkon te një kulm i pa vizituar.
4. Shto atë kulm dhe brinjë në MST.
5. Përsërit derisa të gjitha kulmet të jenë të përfshirë.

**Struktura kryesore:** **Radhë me Prioritet (Min-Heap)** — nxjerr me efikasitet brinjën me peshë minimale.

**Primi funksionon më mirë** kur grafi është **i dendur** (shumë brinjë).

---

### Kruskal kundrejt Prim — Krahasimi

| Karakteristika | Kruskal | Prim |
|----------------|---------|------|
| Qasja | Bazuar në brinjë | Bazuar në kulme |
| Punon me grafe jo të lidhur? | Po | Kërkon graf të lidhur |
| Më i mirë për | Grafe të rralla | Grafe të dendura |
| Struktura kryesore | Union-Find (DSU) | Radhë me Prioritet |
| Rezultati | E njëjta peshë MST | E njëjta peshë MST |

---

### Zbatimet në Botën Reale të MST

- **Projektimi i Rrjetit Elektrik** — lidhja e qyteteve me rrjetin elektrik me kabllo minimale
- **Telekomunikacioni** — shtrirja e kabllove me fibër optike midis qendrave të të dhënave me kosto minimale
- **Planifikimi i Rrjetit Rrugor** — rrjeti rrugor me kosto minimale që lidh të gjitha vendbanimet
- **Paraqitja e Qarkut të Integruar** — minimizimi i gjatësisë së telit në projektimin e çipeve VLSI
- **Bioinformatika** — ndërtimi i pemëve filogjenetike (pemët evolucionare të llojeve)
- **Robotikë** — planifikimi i rrugës së eksplorimit me kosto minimale për robot
- **Zhvillimi i Lojërave** — gjenerimi i relievit realist rastësor dhe hartave të dungeon-it

**Si inxhinier do të përdorësh MST kur:**
- Planifikon infrastrukturë që minimizon koston e ndërtimit
- Projekton rrjete komunikimi me eliminimin e tepricës
- Grupon pikat e të dhënave në mësim makine (single-linkage clustering)
- Ndërton protokolle rrugëzimi në rrjetet kompjuterike

---

### Çfarë bën ky ushtrim

Vizualizuesi më poshtë të lejon të:
1. Kalon në modalitetin **Sandbox** dhe **klikon** për të shtuar nyje, pastaj **klikon dy nyje** për të shtuar brinjë (pesha caktohet automatikisht).
2. Shtyp **Kruskal** ose **Prim** për të parë MST-n duke u ndërtuar hap pas hapi.
3. Përdor **Next / Prev** për të rishikuar çdo hap, ose **Auto** për luajtje të vazhdueshme.

- **Brinjët cian** = të pranuara në MST
- **Brinjët e kuqe** = të refuzuara (do të krijonin cikël — vetëm Kruskal)


## Ushtrimi 3 — Pema Përfshirëse Minimum: Kruskal dhe Prim


````python
import tkinter as tk
import random
import math

# =========================
# DATA STRUCTURES
# =========================

class Node:
    def __init__(self, nid, x, y):
        self.id = nid
        self.x = x
        self.y = y


class Edge:
    def __init__(self, u, v, w):
        self.u = u
        self.v = v
        self.w = w
        self.highlight = None


# =========================
# DSU
# =========================

class DSU:
    def __init__(self, nodes):
        self.parent = {n: n for n in nodes}
        self.rank = {n: 0 for n in nodes}

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, a, b):
        ra, rb = self.find(a), self.find(b)
        if ra == rb:
            return False
        if self.rank[ra] < self.rank[rb]:
            self.parent[ra] = rb
        elif self.rank[ra] > self.rank[rb]:
            self.parent[rb] = ra
        else:
            self.parent[rb] = ra
            self.rank[ra] += 1
        return True


# =========================
# MAIN APP
# =========================

class MSTApp:
    def __init__(self, root):
        self.root = root
        self.root.title("MST Visualizer")

        self.canvas = tk.Canvas(root, bg="#111")
        self.canvas.pack(fill="both", expand=True)

        self.nodes = []
        self.edges = []

        self.mode = "menu"
        self.selected_node = None

        # Step system
        self.step_index = 0
        self.execution_mode = "manual"
        self.current_algorithm = None

        self.build_ui()
        self.bind_events()

    # =========================
    # UI
    # =========================

    def build_ui(self):
        frame = tk.Frame(self.root, bg="#222")
        frame.place(x=10, y=10)

        tk.Button(frame, text="Prim", command=self.start_prim).pack(side="left")
        tk.Button(frame, text="Kruskal", command=self.start_kruskal).pack(side="left")
        tk.Button(frame, text="Sandbox", command=self.start_sandbox).pack(side="left")

        tk.Button(frame, text="Next", command=self.next_step).pack(side="left")
        tk.Button(frame, text="Prev", command=self.prev_step).pack(side="left")

        tk.Button(frame, text="Auto", command=self.toggle_auto).pack(side="left")
        tk.Button(frame, text="Reset", command=self.reset).pack(side="left")

        # Still allow exit via button
        tk.Button(frame, text="Exit", command=self.root.destroy).pack(side="left")

        self.status = tk.Label(self.root, fg="white", bg="#111")
        self.status.place(x=10, y=50)

    def bind_events(self):
        self.canvas.bind("<Button-1>", self.on_click)
        self.canvas.bind("<B1-Motion>", self.drag_node)

        self.root.bind("n", lambda e: self.next_step())
        self.root.bind("p", lambda e: self.prev_step())
        self.root.bind("<space>", lambda e: self.toggle_auto())

        # ESC to exit (important since no window controls)
        self.root.bind("<Escape>", lambda e: self.root.destroy())

    # =========================
    # GRAPH
    # =========================

    def add_node(self, x, y):
        self.nodes.append(Node(len(self.nodes), x, y))
        self.draw()

    def get_node(self, x, y, r=15):
        for n in self.nodes:
            if math.hypot(n.x - x, n.y - y) < r:
                return n
        return None

    def add_edge(self, u, v):
        if u.id != v.id:
            self.edges.append(Edge(u.id, v.id, random.randint(1, 20)))
        self.draw()

    # =========================
    # SANDBOX
    # =========================

    def start_sandbox(self):
        self.mode = "sandbox"

    def on_click(self, event):
        if self.mode != "sandbox":
            return

        n = self.get_node(event.x, event.y)

        if not n:
            self.add_node(event.x, event.y)
            return

        if not self.selected_node:
            self.selected_node = n
        else:
            self.add_edge(self.selected_node, n)
            self.selected_node = None

    def drag_node(self, event):
        if self.mode != "sandbox":
            return
        n = self.get_node(event.x, event.y)
        if n:
            n.x, n.y = event.x, event.y
            self.draw()

    # =========================
    # RESET
    # =========================

    def reset(self):
        self.nodes = []
        self.edges = []
        self.step_index = 0
        self.current_algorithm = None
        self.execution_mode = "manual"
        self.draw()

    # =========================
    # STEP CONTROL
    # =========================

    def next_step(self):
        self.step_index += 1
        self.replay()

    def prev_step(self):
        if self.step_index > 0:
            self.step_index -= 1
            self.replay()

    # =========================
    # AUTO MODE
    # =========================

    def toggle_auto(self):
        self.execution_mode = "auto" if self.execution_mode == "manual" else "manual"
        if self.execution_mode == "auto":
            self.auto_loop()

    def auto_loop(self):
        if self.execution_mode == "auto":
            self.next_step()
            self.root.after(400, self.auto_loop)

    # =========================
    # ALGORITHMS
    # =========================

    def start_prim(self):
        self.current_algorithm = "prim"
        self.step_index = 0
        self.replay()

    def start_kruskal(self):
        self.current_algorithm = "kruskal"
        self.step_index = 0
        self.replay()

    def replay(self):
        self.clear_highlights()

        if self.current_algorithm == "prim":
            self.replay_prim(self.step_index)
        elif self.current_algorithm == "kruskal":
            self.replay_kruskal(self.step_index)

        self.draw()
        self.update_status()

    def replay_prim(self, steps):
        if not self.nodes:
            return

        visited = {0}

        for _ in range(steps):
            best = None
            for e in self.edges:
                if (e.u in visited) ^ (e.v in visited):
                    if best is None or e.w < best.w:
                        best = e

            if best:
                best.highlight = "mst"
                visited.add(best.u)
                visited.add(best.v)

    def replay_kruskal(self, steps):
        dsu = DSU(range(len(self.nodes)))
        sorted_edges = sorted(self.edges, key=lambda e: e.w)

        for i in range(min(steps, len(sorted_edges))):
            e = sorted_edges[i]
            if dsu.find(e.u) != dsu.find(e.v):
                dsu.union(e.u, e.v)
                e.highlight = "mst"
            else:
                e.highlight = "reject"

    # =========================
    # DRAW
    # =========================

    def clear_highlights(self):
        for e in self.edges:
            e.highlight = None

    def draw(self):
        self.canvas.delete("all")

        for e in self.edges:
            u = self.nodes[e.u]
            v = self.nodes[e.v]

            color = "#555"
            if e.highlight == "mst":
                color = "#00f5ff"
            elif e.highlight == "reject":
                color = "red"

            self.canvas.create_line(u.x, u.y, v.x, v.y, fill=color, width=2)
            self.canvas.create_text((u.x+v.x)/2, (u.y+v.y)/2,
                                    text=str(e.w), fill="white")

        for n in self.nodes:
            self.canvas.create_oval(n.x-15, n.y-15, n.x+15, n.y+15,
                                    fill="#2ecc71")
            self.canvas.create_text(n.x, n.y, text=n.id, fill="black")

    # =========================
    # STATUS
    # =========================

    def update_status(self):
        self.status.config(
            text=f"{self.current_algorithm} | Step {self.step_index} | Mode: {self.execution_mode}"
        )


# =========================
# RUN
# =========================

if __name__ == "__main__":
    root = tk.Tk()
    app = MSTApp(root)
    root.mainloop()
````

## Ushtrimi 4 — Komponentet e Lidhura Fort

### Cfare jane Komponentet e Lidhura Fort?

Ne nje **graf te drejtuar**, nje **Komponent i Lidhur Fort (SCC — Strongly Connected Component)** eshte nje bashkesi maksimale kulmesh e tille qe:

> Nga cdo kulm i bashkesise mund te arrihet cdo kulm tjeter i asaj bashkesie duke ndjekur drejtimin e brinjeve.

Me fjale te tjera: brenda nje SCC-i, cdo dy kulme jane te lidhur **reciprokisht** — nese mund te shkosh nga **u** te **v**, mund te kthehesh edhe nga **v** te **u**.

---

### Shembull intuitiv

Imagino qytete dhe rruge njekahere. Dy qytete i perkasin te njejit SCC nese mund te udhetosh nga njeri te tjetri **dhe** te kthehesh mbrapsht duke ndjekur gjithmone drejtimin e rrugeve.

```
  [0] --> [1] --> [3] <--> [4] --> [5]
   ^       |
   +---- [2] <--+       [6] <--> [7]

  SCC-A = {0, 1, 2}  (cikel 0->1->2->0)
  SCC-B = {3, 4}     (cikel 3<->4)
  SCC-C = {5}        (nyje e vetme, asnje cikel)
  SCC-D = {6, 7}     (cikel 6<->7)
```

Brinjet **nder-SCC** jane gjithmone njekahere: nuk ekziston cikel qe kalon dy SCC te ndryshme.

---

### Pronat kryesore

| Prona | Shpjegimi |
|-------|-----------|
| **Maksimaliteti** | Nuk mund te shtohet asnje kulm tjeter pa e prishur vetine |
| **Uniciteti** | Cdo kulm i perket saktesisht nje SCC |
| **SCC singleton** | Cdo kulm pa cikel eshte SCC me vetem ate kulm |
| **Graf i pa drejtuar** | Cdo komponent i lidhur eshte automatikisht SCC |

---

### Zbatimet ne boten reale

- **Rrjetet sociale** — gjetja e grupeve me ndjekje reciproke;
- **Kompilatoret** — zbulimi i varësive ciklike ne module kodi;
- **Web crawlers** — grupet e faqeve te nderlidhura;
- **Navigacioni** — verifikimi i qarkullimit dykahesh ne rrjet rrugor.

---

### Cfare ben ky ushtrim

Ky eshte nje **ndertues interaktiv**: ndertoni grafin tuaj te drejtuar nga zeroja, zgjidhni cfaredo bashkesie kulmesh dhe analizoni nese ato formojne komponent te lidhur fort.

**Rendi i punes:**
1. **Nderto Graf** — kliko ne hapesire bosh per te shtuar kulme; kliko kulm A pastaj kulm B per te shtuar brinjen A→B; drag per te levizur kulmet; klik i djathte anon brinjen ne ndertim.
2. **Zgjidh Kulmet** — kliko kulmet qe deshiron te testosh (shfaqen me unaze portokalli).
3. **Analizo SCC** — kodi verifikon nese kulmet e zgjedhura formojne SCC dhe shpjegon pse po ose pse jo.

Provo hipoteza te ndryshme: zgjedh kulme qe formojne cikel, pastaj shto nje kulm qe nuk perfshin cikel — shiko si ndryshon rezultati!

````python
import sys
import pygame
import math
from collections import deque

pygame.init()

info          = pygame.display.Info()
WIDTH, HEIGHT = info.current_w, info.current_h
screen        = pygame.display.set_mode((WIDTH, HEIGHT), pygame.FULLSCREEN)
pygame.display.set_caption("Komponentet e Lidhura Fort -- SCC Vizualizues")

FONT_SM = pygame.font.SysFont("Segoe UI", 13)
FONT_MD = pygame.font.SysFont("Segoe UI", 16)
FONT_LG = pygame.font.SysFont("Segoe UI", 19, bold=True)

# =============================================================================
# COLOUR PALETTE
# =============================================================================
BG          = (13,  17,  30)
UI_BG       = (20,  26,  46)
UI_LINE     = (55,  75, 130)

NODE_R      = 22

C_NODE      = (65,  85, 135)
C_EDGE      = (65,  80, 115)
C_SEL_RING  = (230, 170,  30)
C_SEL_NODE  = (200, 145,  20)
C_PENDING   = (140, 140, 255)

C_SCC_NODE  = ( 45, 185,  80)
C_SCC_EDGE  = ( 45, 185,  80)
C_FAIL_NODE = (200,  55,  55)
C_MISS_EDGE = (200,  55,  55)

C_OK        = ( 60, 210, 100)
C_ERR       = (220,  75,  75)
C_INFO      = (175, 195, 255)
C_DIM       = ( 95, 110, 150)

BTN_NORMAL  = ( 50,  70, 115)
BTN_ACTIVE  = ( 35, 115,  70)
BTN_HOVER   = ( 75, 105, 165)
BTN_DANGER  = (120,  40,  40)
BTN_TEXT    = (255, 255, 255)

PANEL_W     = 370


def quit_game():
    pygame.quit()
    sys.exit()


# =============================================================================
# CURVED ARROW  (same bezier style as Exercise 2)
# =============================================================================
def draw_arrow(surf, color, start, end, width=2, head=13, rad=0.18):
    dx, dy = end[0] - start[0], end[1] - start[1]
    if math.hypot(dx, dy) < 1:
        return
    mx = (start[0] + end[0]) / 2 + rad * (-dy)
    my = (start[1] + end[1]) / 2 + rad * (dx)
    pts = [
        (
            (1 - t/20)**2 * start[0] + 2*(1-t/20)*(t/20)*mx + (t/20)**2*end[0],
            (1 - t/20)**2 * start[1] + 2*(1-t/20)*(t/20)*my + (t/20)**2*end[1],
        )
        for t in range(21)
    ]
    if len(pts) > 1:
        pygame.draw.lines(surf, color, False,
                          [(int(x), int(y)) for x, y in pts], width)

    def draw_head(tx, ty, adx, ady):
        left  = (tx - head*adx + head*0.4*(-ady),
                 ty - head*ady + head*0.4*(adx))
        right = (tx - head*adx - head*0.4*(-ady),
                 ty - head*ady - head*0.4*(adx))
        pygame.draw.polygon(surf, color, [(tx, ty), left, right])

    for i in range(len(pts) - 1, 0, -1):
        px, py = pts[i]
        if math.hypot(px - end[0], py - end[1]) >= NODE_R + 4:
            nx, ny = pts[i - 1]
            adx, ady = px - nx, py - ny
            length = math.hypot(adx, ady)
            if length < 0.1:
                break
            draw_head(px, py, adx / length, ady / length)
            break


def draw_dashed_arrow(surf, color, start, end, rad=-0.28):
    # Dashed curved arrow curving in the OPPOSITE direction to draw_arrow,
    # used to indicate a missing directed path between two nodes.
    dx, dy = end[0] - start[0], end[1] - start[1]
    if math.hypot(dx, dy) < 1:
        return
    mx = (start[0] + end[0]) / 2 + rad * (-dy)
    my = (start[1] + end[1]) / 2 + rad * (dx)
    pts = [
        (
            (1 - t/40)**2 * start[0] + 2*(1-t/40)*(t/40)*mx + (t/40)**2*end[0],
            (1 - t/40)**2 * start[1] + 2*(1-t/40)*(t/40)*my + (t/40)**2*end[1],
        )
        for t in range(41)
    ]
    dash, gap, acc, draw_on = 7, 5, 0, True
    for i in range(1, len(pts)):
        seg = math.hypot(pts[i][0] - pts[i-1][0], pts[i][1] - pts[i-1][1])
        acc += seg
        if draw_on and acc >= dash:
            draw_on, acc = False, 0
        elif not draw_on and acc >= gap:
            draw_on, acc = True, 0
        if draw_on:
            pygame.draw.line(surf, color,
                             (int(pts[i-1][0]), int(pts[i-1][1])),
                             (int(pts[i][0]),   int(pts[i][1])), 2)


# =============================================================================
# BFS REACHABILITY
# =============================================================================
def bfs_reach(adj, src):
    visited = {src}
    q = deque([src])
    while q:
        v = q.popleft()
        for w in adj.get(v, []):
            if w not in visited:
                visited.add(w)
                q.append(w)
    return visited


# =============================================================================
# SCC ANALYSIS  (no Tarjan/Kosaraju -- only reachability checks)
# =============================================================================
def analyze_scc(adj, selected):
    sel = sorted(selected)

    if not sel:
        return None, ["Nuk ka asnje kulm te zgjedhur."], []

    if len(sel) == 1:
        v = sel[0]
        return True, [
            f"Vetem nje kulm zgjedhur: {v}.",
            "",
            "Nje kulm i vetem eshte",
            "gjithmone SCC (rast trivial).",
        ], []

    # BFS from every selected node across the full graph
    reach = {v: bfs_reach(adj, v) for v in sel}

    # Pairs (u, v) where no directed path u -> v exists
    missing = [(u, v) for u in sel for v in sel
               if u != v and v not in reach[u]]

    is_scc = not missing

    # ---- SCC confirmed -------------------------------------------------------
    if is_scc:
        lines = [
            "REZULTATI:  SCC KONFIRMUAR!",
            "",
            f"Kulmet {sel} formojne",
            "nje Komponent te Lidhur Fort.",
            "",
            "Nga cdo kulm arrihen te gjithe",
            "kulmet e tjere te grupit:",
        ]
        for u in sel:
            others = sorted(v for v in sel if v != u)
            lines.append(f"  {u}  ->  {others}  (OK)")
        lines += [
            "",
            "Cikli eshte i mbyllur --",
            "lidhshmeria eshte reciproke.",
        ]
        print("\n[ANALYZE] " + "  ".join(lines[:3]))
        return True, lines, []

    # ---- NOT an SCC ----------------------------------------------------------
    lines = [
        "REZULTATI:  JO SCC!",
        "",
        "Shtegjet qe mungojne:",
    ]

    cant = {}
    for u, v in missing:
        cant.setdefault(u, []).append(v)
    for u in sorted(cant):
        lines.append(f"  {u}  -/>  {sorted(cant[u])}")

    lines.append("")

    # Classify pairs
    oneway, noconn = [], []
    seen_pairs = set()
    for u in sel:
        for v in sel:
            if u >= v or (u, v) in seen_pairs:
                continue
            seen_pairs.add((u, v))
            u2v = v in reach[u]
            v2u = u in reach[v]
            if u2v and not v2u:
                oneway.append(f"  {u}->{v} njekahore (nuk ka {v}->{u})")
            elif v2u and not u2v:
                oneway.append(f"  {v}->{u} njekahore (nuk ka {u}->{v})")
            elif not u2v and not v2u:
                noconn.append(f"  {u} dhe {v}: asnje drejtim")

    if oneway:
        lines.append("Lidhje njekahore (jo ciklike):")
        lines += oneway
        lines.append("")
    if noconn:
        lines.append("Pa lidhje fare:")
        lines += noconn
        lines.append("")

    lines += [
        "Shto brinje per te mbyllur",
        "ciklet midis kulmeve.",
    ]

    print(f"\n[ANALYZE] selected={sel}  missing={missing}")
    return False, lines, missing


# =============================================================================
# BUTTON WIDGET
# =============================================================================
class Button:
    def __init__(self, rect, label, color=None, active_color=None):
        self.rect         = pygame.Rect(rect)
        self.label        = label
        self.color        = color        or BTN_NORMAL
        self.active_color = active_color or BTN_ACTIVE
        self.is_active    = False

    def draw(self, surf):
        mouse = pygame.mouse.get_pos()
        col = (self.active_color if self.is_active
               else BTN_HOVER    if self.rect.collidepoint(mouse)
               else self.color)
        pygame.draw.rect(surf, col,     self.rect, border_radius=7)
        pygame.draw.rect(surf, UI_LINE, self.rect, 2, border_radius=7)
        lbl = FONT_MD.render(self.label, True, BTN_TEXT)
        surf.blit(lbl, lbl.get_rect(center=self.rect.center))

    def hit(self, pos):
        return self.rect.collidepoint(pos)


# =============================================================================
# MAIN APP
# =============================================================================
class App:
    def __init__(self):
        self.nodes     = []        # list of (x, y)
        self.edges     = []        # list of (u, v)  directed
        self.history   = []        # for undo
        self.mode      = "build"   # "build" | "select"
        self.edge_from = None      # first node while drawing an edge
        self.selected  = set()     # selected node ids
        self.result_scc = None     # True | False | None
        self.res_lines  = []
        self.missing    = []

        BX = WIDTH - PANEL_W + 18
        BW = PANEL_W - 36
        BH = 40

        self.btn_build   = Button((BX,  56, BW, BH), "Modo: Nderto Graf")
        self.btn_select  = Button((BX, 106, BW, BH), "Modo: Zgjidh Kulmet")
        self.btn_analyze = Button((BX, 170, BW, BH), "Analizo SCC",
                                  color=(40, 110, 70), active_color=(40, 160, 85))
        self.btn_clrsel  = Button((BX, 220, BW, BH), "Pastro Seleksionin",
                                  color=(100, 80, 25))
        self.btn_undo    = Button((BX, 280, BW, BH), "Undo  [Z]")
        self.btn_reset   = Button((BX, 330, BW, BH), "Reset Graf  [C]",
                                  color=BTN_DANGER)
        self._sync_mode_btns()

    # -- adjacency -------------------------------------------------------------

    def _adj(self):
        adj = {i: [] for i in range(len(self.nodes))}
        for u, v in self.edges:
            adj[u].append(v)
        return adj

    # -- mutations -------------------------------------------------------------

    def add_node(self, pos):
        nid = len(self.nodes)
        self.nodes.append(pos)
        self.history.append(("node", nid))
        self._clear_result()
        print(f"[NODE] id={nid}")

    def add_edge(self, u, v):
        if u == v or (u, v) in self.edges:
            return
        self.edges.append((u, v))
        self.history.append(("edge", u, v))
        self._clear_result()
        print(f"[EDGE] {u} -> {v}")

    def undo(self):
        if not self.history:
            return
        act = self.history.pop()
        if act[0] == "node":
            nid = act[1]
            self.nodes.pop()
            self.edges    = [(u, v) for u, v in self.edges
                             if u != nid and v != nid]
            self.selected.discard(nid)
        elif act[0] == "edge":
            self.edges.remove((act[1], act[2]))
        self.edge_from = None
        self._clear_result()

    def reset(self):
        self.nodes, self.edges, self.history = [], [], []
        self.selected  = set()
        self.edge_from = None
        self._clear_result()
        print("[RESET]")

    def _clear_result(self):
        self.result_scc = None
        self.res_lines  = []
        self.missing    = []

    # -- mode ------------------------------------------------------------------

    def _sync_mode_btns(self):
        self.btn_build.is_active  = (self.mode == "build")
        self.btn_select.is_active = (self.mode == "select")

    def set_mode(self, m):
        self.mode = m
        self.edge_from = None
        self._sync_mode_btns()

    # -- hit-test --------------------------------------------------------------

    def node_at(self, pos):
        for i, (x, y) in enumerate(self.nodes):
            if math.hypot(pos[0] - x, pos[1] - y) < NODE_R + 5:
                return i
        return None

    # -- click -----------------------------------------------------------------

    def handle_click(self, pos):
        if pos[0] >= WIDTH - PANEL_W:
            if   self.btn_build.hit(pos):   self.set_mode("build")
            elif self.btn_select.hit(pos):  self.set_mode("select")
            elif self.btn_analyze.hit(pos): self._run_analyze()
            elif self.btn_clrsel.hit(pos):
                self.selected = set()
                self._clear_result()
            elif self.btn_undo.hit(pos):    self.undo()
            elif self.btn_reset.hit(pos):   self.reset()
            return

        nid = self.node_at(pos)

        if self.mode == "build":
            if nid is None:
                self.add_node(pos)
                self.edge_from = None
            elif self.edge_from is None:
                self.edge_from = nid
            else:
                self.add_edge(self.edge_from, nid)
                self.edge_from = None

        elif self.mode == "select":
            if nid is not None:
                if nid in self.selected:
                    self.selected.discard(nid)
                else:
                    self.selected.add(nid)
                self._clear_result()

    def _run_analyze(self):
        if not self.nodes:
            self.res_lines = ["Shto te pakten dy nyje!"]
            return
        adj = self._adj()
        is_scc, lines, missing = analyze_scc(adj, self.selected)
        self.result_scc = is_scc
        self.res_lines  = lines
        self.missing    = missing

    # -- draw ------------------------------------------------------------------

    def draw(self):
        screen.fill(BG)
        canvas_x  = WIDTH - PANEL_W
        sel       = self.selected
        confirmed = self.result_scc is True
        denied    = self.result_scc is False

        # Edges
        for u, v in self.edges:
            if confirmed and u in sel and v in sel:
                color, width = C_SCC_EDGE, 3
            else:
                color, width = C_EDGE, 2
            draw_arrow(screen, color, self.nodes[u], self.nodes[v], width)

        # Missing-path dashed arcs
        if denied:
            for u, v in self.missing:
                if u < len(self.nodes) and v < len(self.nodes):
                    draw_dashed_arrow(screen, C_MISS_EDGE,
                                      self.nodes[u], self.nodes[v])

        # Pending edge preview while drawing
        if self.mode == "build" and self.edge_from is not None:
            mx, my = pygame.mouse.get_pos()
            draw_arrow(screen, C_PENDING,
                       self.nodes[self.edge_from], (mx, my), 2)

        # Nodes
        for i, (x, y) in enumerate(self.nodes):
            in_sel = i in sel

            if in_sel:
                ring = (C_SCC_NODE  if confirmed
                        else C_FAIL_NODE if denied
                        else C_SEL_RING)
                pygame.draw.circle(screen, ring, (x, y), NODE_R + 6)

            fill = (C_SCC_NODE  if (confirmed and in_sel)
                    else C_FAIL_NODE if (denied    and in_sel)
                    else C_SEL_NODE  if in_sel
                    else C_PENDING   if i == self.edge_from
                    else C_NODE)

            pygame.draw.circle(screen, fill, (x, y), NODE_R)
            lbl = FONT_MD.render(str(i), True, (255, 255, 255))
            screen.blit(lbl, lbl.get_rect(center=(x, y)))

        pygame.draw.line(screen, UI_LINE, (canvas_x, 0), (canvas_x, HEIGHT), 2)
        self._draw_panel(canvas_x)
        pygame.display.flip()

    def _draw_panel(self, px):
        pygame.draw.rect(screen, UI_BG, (px, 0, PANEL_W, HEIGHT))
        tx = px + 16

        def t(s, y, color=C_INFO, font=FONT_MD):
            screen.blit(font.render(s, True, color), (tx, y))

        t("SCC  --  Graf Interaktiv", 12, color=(215, 225, 255), font=FONT_LG)
        pygame.draw.line(screen, UI_LINE, (px+8, 44), (WIDTH-8, 44), 1)

        for btn in (self.btn_build, self.btn_select, self.btn_analyze,
                    self.btn_clrsel, self.btn_undo, self.btn_reset):
            btn.draw(screen)

        y = 390
        pygame.draw.line(screen, UI_LINE, (px+8, y), (WIDTH-8, y), 1)
        y += 8

        mode_lbl = "Nderto Graf" if self.mode == "build" else "Zgjidh Kulmet"
        t(f"Modo: {mode_lbl}", y, C_DIM);  y += 20
        t(f"Kulme: {len(self.nodes)}    Brinje: {len(self.edges)}", y, C_DIM); y += 20
        sel_str = str(sorted(self.selected)) if self.selected else "asnje"
        t(f"Zgjedhur: {sel_str}", y,
          C_SEL_RING if self.selected else C_DIM);  y += 28

        if self.res_lines:
            pygame.draw.line(screen, UI_LINE, (px+8, y), (WIDTH-8, y), 1)
            y += 8
            res_col = (C_OK  if self.result_scc is True
                       else C_ERR if self.result_scc is False
                       else C_INFO)
            for ln in self.res_lines:
                if y > HEIGHT - 90:
                    t("...", y, C_DIM, FONT_SM)
                    break
                t(ln, y, res_col, FONT_SM);  y += 17

        y = HEIGHT - 92
        pygame.draw.line(screen, UI_LINE, (px+8, y), (WIDTH-8, y), 1)
        y += 6
        for hint in ["Klik boshlleku -> shto nyje",
                     "Klik nyje->nyje -> brinje (build)",
                     "Klik nyje -> sel/desel (select)",
                     "Z=Undo  C=Reset  ESC=Dil"]:
            t(hint, y, C_DIM, FONT_SM);  y += 17

    # -- main loop -------------------------------------------------------------

    def run(self):
        clock = pygame.time.Clock()
        print("\n" + "="*60)
        print("  SCC Vizualizues  --  shtyp ESC per te dale")
        print("="*60)
        while True:
            clock.tick(60)
            for event in pygame.event.get():
                if event.type == pygame.QUIT:
                    quit_game()
                if event.type == pygame.KEYDOWN:
                    if   event.key == pygame.K_ESCAPE: quit_game()
                    elif event.key == pygame.K_z:      self.undo()
                    elif event.key == pygame.K_c:      self.reset()
                    elif event.key == pygame.K_b:      self.set_mode("build")
                    elif event.key == pygame.K_s:      self.set_mode("select")
                if event.type == pygame.MOUSEBUTTONDOWN:
                    self.handle_click(event.pos)
            self.draw()


if __name__ == "__main__":
    App().run()
````

## Ushtrimi 5 — Renditjet Topologjike

### Cfare eshte Renditja Topologjike?

**Renditja Topologjike** e nje **grafi te drejtuar acilik (DAG — Directed Acyclic Graph)** eshte nje renditje lineare e te gjitha kulmeve e tille qe:

> Per cdo brinje te drejtuar **u -> v**, kulmi **u** shfaqet **para** kulmin **v** ne renditje.

Me fjale te tjera: nese ka nje varesi "u duhet te behet para v", atehere u duhet te vije i pari.

---

### Kushti i domosdoshem: grafu nuk duhet te kete cikle

Renditja topologjike **ekziston vetem per DAG**. Nese grafi ka cikel (p.sh. A->B->C->A), eshte e pamundur te rendisesh kulmet pa kontradikter — A duhet para B, B para C, dhe C para A, gje e pamundur.

---

### Shembull intuitiv: varesitë e lendeve

Imagino se deshiron te regjistron lende ne universitet, ku disa lende kerkojne parakushte:

```
MAT1 --> MAT2 --> ALG1 --> DB --> NET
  |                         ^      ^
  +--> PRG1 --> PRG2 --> OS-+      |
                 +-------------------+
```

Cdo renditje e vlefshme (p.sh. MAT1 -> PRG1 -> MAT2 -> PRG2 -> ALG1 -> OS -> DB -> NET)
respekton te gjitha varesitë.  
Mund te kete **shume renditje te vlefshme** — renditja topologjike nuk eshte unike.

---

### Pronat kryesore

| Prona | Shpjegimi |
|-------|-----------|
| **Ekzistenca** | Vetem per DAG (grafet pa cikle) |
| **Uniciteti** | Zakonisht nuk eshte unike — mund te kete shume renditje te vlefshme |
| **Nyjet "te gatshme"** | Nyjet me in-degree 0 (pa parakushte) mund te vendosen te parat |
| **Efekti zinxhir** | Pasi shtohet nje nyje, mund te lirohen nyje te tjera me in-degree 0 |

---

### Zbatimet ne boten reale

- **Sistemet e ndertimit te kodit** (Make, Gradle, CMake) — kompilimi i moduleve ne rendin e duhur;
- **Plani mesimor** — renditja e lendeve sipas parakushteve;
- **Menaxhimi i projekteve** — planifikimi i detyrave me varesi;
- **Spreadsheet-et** — llogaritja e formulave ne rendin e sakte (Excel, Google Sheets);
- **Paketat e software-it** — instalimi i varesive (npm, pip, apt).

---

### Cfare ben ky ushtrim

Vizualizuesi paraqet nje DAG te lendeve universitare me varesi mes tyre.  
**Detyra juaj:** klikoni lendet ne rendin e duhur topologjik.

- **Nyjet e gjelbra te ndricuara** = mund te zgjidhen tani (te gjitha parakushtet jane plotesuar)
- **Nyjet e errëta** = nuk mund te zgjidhen ende (ka parakushte te paperfunduara)
- **Nyjet e gjelbra te plota** = tashmë te shtuara ne renditje

Verejtje: provoni te klikoni renditje te ndryshme — do te shihni se ka shume renditje te vlefshme!


## Ushtrimi 5 — Renditjet Topologjike


````python
import tkinter as tk
import math

# =============================================================================
# GRAPH DATA — University course prerequisite DAG
#
# COURSES  : {id: (short_code, full_name)}
# NODE_POS : base canvas positions, scaled to window at runtime
# PREREQS  : (u, v) means course u must be completed before course v
# =============================================================================

COURSES = {
    0: ("MAT1", "Matematike I"),
    1: ("MAT2", "Matematike II"),
    2: ("ALG1", "Algoritmet"),
    3: ("PRG1", "Programim I"),
    4: ("PRG2", "Programim II"),
    5: ("DB",   "Baza te Dhenave"),
    6: ("OS",   "Sist. Operative"),
    7: ("NET",  "Rrjeta"),
}

BASE_W, BASE_H = 920, 460

NODE_POS = {
    0: (120, 230),
    1: (280, 130),
    2: (460, 130),
    3: (280, 330),
    4: (460, 330),
    5: (640, 230),
    6: (640, 130),
    7: (800, 230),
}

PREREQS = [
    (0, 1),   # MAT1  -> MAT2
    (0, 3),   # MAT1  -> PRG1
    (1, 2),   # MAT2  -> ALG1
    (3, 4),   # PRG1  -> PRG2
    (2, 5),   # ALG1  -> DB
    (4, 5),   # PRG2  -> DB
    (4, 6),   # PRG2  -> OS
    (5, 7),   # DB    -> NET
    (6, 7),   # OS    -> NET
]

NODE_R = 30


# =============================================================================
# VISUALISER
# =============================================================================

class TopoVisualizer:
    def __init__(self, root):
        self.root  = root
        self.order = []     # topological order built by the student
        self.done  = set()  # ids already placed

        root.title("Renditjet Topologjike")
        root.configure(bg="#111")
        root.attributes("-fullscreen", True)
        root.bind("<Escape>", lambda e: root.destroy())
        root.protocol("WM_DELETE_WINDOW", root.destroy)

        tk.Label(root,
                 text="Renditjet Topologjike — Vizualizues",
                 font=("Helvetica", 20, "bold"), bg="#111", fg="white"
                 ).pack(pady=(16, 2))
        tk.Label(root,
                 text="Kliko lendet sipas rendit topologjik. "
                      "Vetem lendet me parakushtet e plota jane te disponueshme (jeshile e ndricuar).",
                 font=("Helvetica", 12), bg="#111", fg="#aaa",
                 wraplength=1000).pack(pady=(0, 6))

        self.canvas = tk.Canvas(root, bg="#1a1a2e", highlightthickness=0)
        self.canvas.pack(fill="both", expand=True, padx=20, pady=(0, 4))
        self.canvas.bind("<Button-1>", self.on_click)

        row = tk.Frame(root, bg="#111")
        row.pack(pady=(2, 4))
        tk.Label(row, text="Rendi juaj: ",
                 font=("Courier New", 13), bg="#111", fg="#aaa").pack(side="left")
        self.order_label = tk.Label(row, text="[ ]",
                                    font=("Courier New", 13, "bold"),
                                    bg="#111", fg="#2ecc71")
        self.order_label.pack(side="left")

        ctrl = tk.Frame(root, bg="#111")
        ctrl.pack(pady=(0, 6))
        kw = dict(font=("Courier New", 13), relief="solid", bd=1)
        tk.Button(ctrl, text="Rifillo",
                  bg="#e74c3c", fg="white",
                  command=self.reset, **kw).pack(side="left", padx=8)
        tk.Button(ctrl, text="Dil",
                  bg="#555", fg="white",
                  command=root.destroy, **kw).pack(side="left", padx=8)

        self.status = tk.Label(
            root,
            text="Kliko nje lende te disponueshme (jeshile e ndricuar) per ta shtuar.",
            font=("Courier New", 12), bg="#111", fg="#aaa")
        self.status.pack(pady=(0, 6))

        tk.Label(root, text="Escape = dil",
                 font=("Helvetica", 9), bg="#111", fg="#333"
                 ).pack(side="bottom", pady=(0, 6))

        root.update_idletasks()
        self.draw()

    # -- helpers --------------------------------------------------------------

    def available(self):
        # nodes whose prerequisites are all already placed
        return [n for n in COURSES
                if n not in self.done
                and all(p in self.done for p, q in PREREQS if q == n)]

    def node_pos(self, nid):
        cw = max(self.canvas.winfo_width(),  1)
        ch = max(self.canvas.winfo_height(), 1)
        bx, by = NODE_POS[nid]
        return bx * cw / BASE_W, by * ch / BASE_H

    def _arrow(self, x1, y1, x2, y2, color, width):
        dx, dy = x2 - x1, y2 - y1
        dist = math.hypot(dx, dy)
        if dist < 1:
            return
        ux, uy = dx / dist, dy / dist
        sx, sy = x1 + ux * NODE_R,        y1 + uy * NODE_R
        ex, ey = x2 - ux * (NODE_R + 14), y2 - uy * (NODE_R + 14)
        self.canvas.create_line(sx, sy, ex, ey,
                                fill=color, width=width,
                                arrow="last", arrowshape=(13, 16, 5))

    # -- draw -----------------------------------------------------------------

    def draw(self):
        self.canvas.delete("all")
        avail = set(self.available())

        for u, v in PREREQS:
            x1, y1 = self.node_pos(u)
            x2, y2 = self.node_pos(v)
            if u in self.done and v in self.done:
                color, width = "#27ae60", 2
            elif u in self.done:
                color, width = "#555", 2
            else:
                color, width = "#3a3a4a", 1
            self._arrow(x1, y1, x2, y2, color, width)

        for nid, (code, name) in COURSES.items():
            x, y = self.node_pos(nid)
            if nid in self.done:
                fill, outline, ow = "#27ae60", "#aaa", 1
            elif nid in avail:
                fill, outline, ow = "#1a5c32", "#2ecc71", 3
            else:
                fill, outline, ow = "#2c2c3e", "#555",    1
            self.canvas.create_oval(x - NODE_R, y - NODE_R,
                                    x + NODE_R, y + NODE_R,
                                    fill=fill, outline=outline, width=ow)
            self.canvas.create_text(x, y - 8, text=code,
                                    font=("Helvetica", 11, "bold"), fill="white")
            self.canvas.create_text(x, y + 9, text=name,
                                    font=("Helvetica", 8), fill="#ccc")

            # sequence-number badge top-right of node
            if nid in self.done:
                idx = self.order.index(nid) + 1
                bx2, by2 = x + NODE_R - 2, y - NODE_R + 2
                self.canvas.create_oval(bx2 - 10, by2 - 10,
                                        bx2 + 10, by2 + 10,
                                        fill="#f39c12", outline="")
                self.canvas.create_text(bx2, by2, text=str(idx),
                                        font=("Helvetica", 9, "bold"), fill="#111")

        # legend — bottom left of canvas
        ch = max(self.canvas.winfo_height(), 1)
        lx, ly = 18, ch - 90
        items = [("E disponueshme — kliko per ta shtuar", "#2ecc71"),
                 ("E shtuar ne renditje",                "#27ae60"),
                 ("Jo e disponueshme ende",              "#555")]
        for i, (txt, col) in enumerate(items):
            self.canvas.create_text(lx, ly + i * 26, text="[  ] " + txt,
                                    anchor="w", fill=col, font=("Helvetica", 11))

        codes = [COURSES[n][0] for n in self.order]
        self.order_label.config(
            text="[ " + "  ->  ".join(codes) + " ]" if codes else "[ ]")

    # -- interaction ----------------------------------------------------------

    def on_click(self, event):
        avail = set(self.available())
        for nid in COURSES:
            if nid in self.done:
                continue
            x, y = self.node_pos(nid)
            if math.hypot(event.x - x, event.y - y) < NODE_R + 6:
                if nid not in avail:
                    missing = [COURSES[p][0] for p, q in PREREQS
                               if q == nid and p not in self.done]
                    self.status.config(
                        fg="#e74c3c",
                        text=f"'{COURSES[nid][0]}' nuk mund te zgjidhet tani! "
                             f"Parakushtet e munguar: {missing}")
                else:
                    self.done.add(nid)
                    self.order.append(nid)
                    if len(self.done) == len(COURSES):
                        self.status.config(
                            fg="#2ecc71",
                            text="Renditja juaj topologjike eshte e plote dhe e sakte!")
                    else:
                        nxt = [COURSES[n][0] for n in self.available()]
                        self.status.config(
                            fg="#aaa",
                            text=f"'{COURSES[nid][0]}' u shtua. "
                                 f"Lendet e disponueshme tani: {nxt}")
                self.draw()
                return

    def reset(self):
        self.order = []
        self.done  = set()
        self.status.config(
            fg="#aaa",
            text="Kliko nje lende te disponueshme (jeshile e ndricuar) per ta shtuar.")
        self.draw()


if __name__ == "__main__":
    root = tk.Tk()
    TopoVisualizer(root)
    root.mainloop()
````

````python
import sys
import math
import time
import random
from collections import deque
import pygame

pygame.init()

info          = pygame.display.Info()
WIDTH, HEIGHT = info.current_w, info.current_h
screen        = pygame.display.set_mode((WIDTH, HEIGHT), pygame.FULLSCREEN)
pygame.display.set_caption("Renditjet Topologjike -- Loje Interaktive")

FONT_XS = pygame.font.SysFont("Segoe UI", 11)
FONT_SM = pygame.font.SysFont("Segoe UI", 13)
FONT_MD = pygame.font.SysFont("Segoe UI", 15)
FONT_LG = pygame.font.SysFont("Segoe UI", 17, bold=True)
FONT_XL = pygame.font.SysFont("Segoe UI", 24, bold=True)

# =============================================================================
# COLOURS
# =============================================================================
BG          = (13,  17,  30)
UI_BG       = (20,  26,  46)
UI_LINE     = (55,  75, 130)
SEQ_BG      = (18,  24,  42)

NODE_R      = 34

C_DEFAULT   = ( 60,  82, 132)
C_AVAIL     = ( 30, 120,  55)
C_PLACED    = ( 45, 175,  75)
C_LOCKED    = ( 30,  37,  60)
C_TYPING    = ( 85, 115, 185)
C_EDGE      = ( 58,  75, 112)
C_EDGE_DONE = ( 40, 155,  68)
C_EDGE_PEND = (135, 135, 255)

C_OK        = ( 55, 205,  95)
C_ERR       = (215,  65,  65)
C_WARN      = (220, 165,  30)
C_INFO      = (175, 195, 255)
C_DIM       = ( 90, 105, 148)
C_WHITE     = (255, 255, 255)

BTN_N  = ( 50,  70, 115)
BTN_A  = ( 35, 115,  70)
BTN_H  = ( 75, 108, 168)
BTN_D  = (115,  38,  38)
BTN_G  = ( 28, 100,  58)

PANEL_W = 375

CONFETTI = [
    (255,  80,  80), ( 80, 200,  80), ( 80, 130, 255),
    (255, 200,  50), (200,  80, 255), ( 80, 220, 220),
    (255, 140,  30), (255, 255,  80),
]

# =============================================================================
# PRESET SCENARIOS
# =============================================================================
PRESETS = {
    "Lende Universitare": {
        "desc": "Cila lende kerkon cilen?",
        "nodes": ["MAT1","PRG1","MAT2","PRG2","ALG","OS","DB","NET"],
        "edges": [
            ("MAT1","MAT2"),("MAT1","PRG1"),("MAT2","ALG"),
            ("PRG1","PRG2"),("ALG","DB"),("PRG2","DB"),
            ("PRG2","OS"),("DB","NET"),("OS","NET"),
        ],
    },
    "Nderto Shtepine": {
        "desc": "Cilin hap para cilit?",
        "nodes": ["Themeli","Muret","Cati","Elektrike","Hidraulike","Suvatim","Lyerja"],
        "edges": [
            ("Themeli","Muret"),("Muret","Cati"),("Muret","Elektrike"),
            ("Muret","Hidraulike"),("Cati","Suvatim"),("Elektrike","Suvatim"),
            ("Hidraulike","Suvatim"),("Suvatim","Lyerja"),
        ],
    },
    "Gatuaj Paste": {
        "desc": "Cili hap vjen para cilit?",
        "nodes": ["Bli paste","Ziej ujen","Shto paste","Bej sosen","Kulloje","Shto sosen","Shperndaj"],
        "edges": [
            ("Bli paste","Ziej ujen"),("Ziej ujen","Shto paste"),
            ("Shto paste","Kulloje"),("Bej sosen","Shto sosen"),
            ("Kulloje","Shto sosen"),("Shto sosen","Shperndaj"),
        ],
    },
}

# =============================================================================
# GRAPH HELPERS
# =============================================================================

def build_predecessors(edges, n):
    pred = {i: [] for i in range(n)}
    for u, v in edges:
        pred[v].append(u)
    return pred


def available_nodes(pred, placed, n):
    return [v for v in range(n)
            if v not in placed
            and all(u in placed for u in pred[v])]


def has_cycle(edges, n):
    adj   = {i: [] for i in range(n)}
    for u, v in edges:
        adj[u].append(v)
    WHITE, GRAY, BLACK = 0, 1, 2
    color = [WHITE] * n

    def dfs(v):
        color[v] = GRAY
        for w in adj[v]:
            if color[w] == GRAY:
                return True
            if color[w] == WHITE and dfs(w):
                return True
        color[v] = BLACK
        return False

    return any(color[v] == WHITE and dfs(v) for v in range(n))


# =============================================================================
# PARTICLE SYSTEM  (confetti on win)
# =============================================================================

class Particle:
    def __init__(self, x, y):
        self.x     = float(x)
        self.y     = float(y)
        self.vx    = random.uniform(-4, 4)
        self.vy    = random.uniform(-10, -3)
        self.color = random.choice(CONFETTI)
        self.w     = random.randint(6, 14)
        self.h     = random.randint(4,  8)
        self.angle = random.uniform(0, 360)
        self.spin  = random.uniform(-6, 6)
        self.life  = 1.0

    def update(self):
        self.vy    += 0.25          # gravity
        self.x     += self.vx
        self.y     += self.vy
        self.angle += self.spin
        self.life  -= 0.009

    def alive(self):
        return self.life > 0 and self.y < HEIGHT + 30

    def draw(self, surf):
        alpha  = max(0, int(self.life * 255))
        col    = tuple(int(c * self.life) for c in self.color)
        s      = pygame.Surface((self.w, self.h), pygame.SRCALPHA)
        s.fill((*col, alpha))
        rot    = pygame.transform.rotate(s, self.angle)
        surf.blit(rot, (int(self.x - rot.get_width()  / 2),
                        int(self.y - rot.get_height() / 2)))


def spawn_confetti(n=120):
    return [Particle(random.randint(0, WIDTH), random.randint(-60, 0))
            for _ in range(n)]


# =============================================================================
# BUTTON WIDGET
# =============================================================================

class Button:
    def __init__(self, rect, label, color=None, active_color=None, font=None):
        self.rect         = pygame.Rect(rect)
        self.label        = label
        self.color        = color        or BTN_N
        self.active_color = active_color or BTN_A
        self.font         = font         or FONT_MD
        self.is_active    = False

    def draw(self, surf):
        mouse = pygame.mouse.get_pos()
        col = (self.active_color if self.is_active
               else BTN_H        if self.rect.collidepoint(mouse)
               else self.color)
        pygame.draw.rect(surf, col,     self.rect, border_radius=7)
        pygame.draw.rect(surf, UI_LINE, self.rect, 2, border_radius=7)
        lbl = self.font.render(self.label, True, C_WHITE)
        surf.blit(lbl, lbl.get_rect(center=self.rect.center))

    def hit(self, pos):
        return self.rect.collidepoint(pos)


# =============================================================================
# CURVED ARROW  (same bezier style as Exercises 2 and 4)
# =============================================================================

def draw_arrow(surf, color, start, end, width=2, head=14, rad=0.18):
    dx, dy = end[0] - start[0], end[1] - start[1]
    if math.hypot(dx, dy) < 1:
        return
    mx = (start[0] + end[0]) / 2 + rad * (-dy)
    my = (start[1] + end[1]) / 2 + rad * (dx)
    pts = [
        (
            (1 - t/20)**2 * start[0] + 2*(1-t/20)*(t/20)*mx + (t/20)**2*end[0],
            (1 - t/20)**2 * start[1] + 2*(1-t/20)*(t/20)*my + (t/20)**2*end[1],
        )
        for t in range(21)
    ]
    if len(pts) > 1:
        pygame.draw.lines(surf, color, False,
                          [(int(x), int(y)) for x, y in pts], width)

    def draw_head(tx, ty, adx, ady):
        lp = (tx - head*adx + head*0.4*(-ady), ty - head*ady + head*0.4*(adx))
        rp = (tx - head*adx - head*0.4*(-ady), ty - head*ady - head*0.4*(adx))
        pygame.draw.polygon(surf, color, [(tx, ty), lp, rp])

    for i in range(len(pts) - 1, 0, -1):
        px, py = pts[i]
        if math.hypot(px - end[0], py - end[1]) >= NODE_R + 5:
            nx, ny = pts[i - 1]
            adx, ady = px - nx, py - ny
            length   = math.hypot(adx, ady)
            if length < 0.1:
                break
            draw_head(px, py, adx / length, ady / length)
            break


# =============================================================================
# NODE
# =============================================================================

class Node:
    MAX_CHARS = 11

    def __init__(self, nid, x, y):
        self.id      = nid
        self.x       = x
        self.y       = y
        self.name    = ""
        self.typing  = True       # True while user is entering the name

        # animation
        self.flash_frames = 0     # >0 = red-flash effect on wrong click
        self.flash_color  = C_ERR

    def display_name(self):
        return self.name if self.name else f"N{self.id}"

    def font_for_name(self):
        n = len(self.display_name())
        if n <= 4:
            return FONT_LG
        if n <= 8:
            return FONT_MD
        return FONT_SM

    def pos(self):
        return (self.x, self.y)


# =============================================================================
# MAIN APPLICATION
# =============================================================================

class App:
    def __init__(self):
        # Graph data
        self.nodes   = []          # list of Node
        self.edges   = []          # list of (u, v)  directed
        self.history = []

        # Build phase state
        self.phase     = "build"   # "build" | "game" | "won"
        self.mode      = "node"    # "node" | "edge"  (build sub-mode)
        self.edge_from = None
        self.typing_node = None    # Node currently being named

        # Game phase state
        self.placed     = set()
        self.order      = []
        self.mistakes   = 0
        self.hint_node  = None
        self.start_ts   = None
        self.elapsed    = 0.0

        # Feedback (wrong click flash)
        self.msg_text   = ""
        self.msg_color  = C_INFO
        self.msg_timer  = 0        # frames remaining

        # Win screen
        self.particles  = []
        self.win_time   = 0.0
        self.stars      = 0

        # Cursor blink
        self.blink_t    = 0

        self._build_buttons()

    # ── button layout ─────────────────────────────────────────────────────────

    def _build_buttons(self):
        BX = WIDTH - PANEL_W + 16
        BW = PANEL_W - 32
        SH = 36           # short button height
        TH = 44           # tall button height

        # Build phase
        self.btn_mode_node  = Button((BX,  56, BW, SH), "Modo: Shto Nyje")
        self.btn_mode_edge  = Button((BX,  98, BW, SH), "Modo: Shto Brinje")

        y = 150
        self.preset_btns = []
        for name in PRESETS:
            short = name[:26]
            self.preset_btns.append(
                Button((BX, y, BW, SH), short, color=(50, 60, 105))
            )
            y += 42

        self.btn_start  = Button((BX, y+8, BW, TH), "FILLO LOJËN  >>",
                                  color=BTN_G, active_color=(40, 145, 80),
                                  font=FONT_LG)

        # Game / won phase
        self.btn_hint    = Button((BX, 56, BW, SH),  "Trego Nje Sugjerim",
                                   color=(70, 55, 110))
        self.btn_restart = Button((BX, 100, BW, SH), "Rifillo Lojën",
                                   color=BTN_D)
        self.btn_back    = Button((BX, 144, BW, SH), "Kthehu tek Ndertimi")

        self._sync_mode_buttons()

    def _sync_mode_buttons(self):
        self.btn_mode_node.is_active = (self.mode == "node")
        self.btn_mode_edge.is_active = (self.mode == "edge")

    # ── adjacency / predecessor helpers ───────────────────────────────────────

    def _pred(self):
        return build_predecessors(self.edges, len(self.nodes))

    def _available(self):
        pred = self._pred()
        return available_nodes(pred, self.placed, len(self.nodes))

    # ── graph mutations ───────────────────────────────────────────────────────

    def _add_node(self, pos):
        nid  = len(self.nodes)
        node = Node(nid, pos[0], pos[1])
        self.nodes.append(node)
        self.history.append(("node", nid))
        self.typing_node = node
        self._clear_msg()
        print(f"[NODE] id={nid}")

    def _confirm_name(self):
        if self.typing_node:
            self.typing_node.typing = False
            print(f"[NAME] node {self.typing_node.id} = '{self.typing_node.display_name()}'")
        self.typing_node = None

    def _add_edge(self, u, v):
        if u == v or (u, v) in self.edges:
            return
        self.edges.append((u, v))
        self.history.append(("edge", u, v))
        self._clear_msg()
        print(f"[EDGE] {u} -> {v}")

    def _undo(self):
        if not self.history:
            return
        self.typing_node = None
        act = self.history.pop()
        if act[0] == "node":
            self.nodes.pop()
            nid = act[1]
            self.edges = [(u, v) for u, v in self.edges
                          if u != nid and v != nid]
        elif act[0] == "edge":
            self.edges.remove((act[1], act[2]))
        self.edge_from = None

    def _reset_graph(self):
        self.nodes, self.edges, self.history = [], [], []
        self.edge_from   = None
        self.typing_node = None
        self.phase = "build"
        self._clear_msg()
        print("[RESET]")

    def _clear_msg(self):
        self.msg_text  = ""
        self.msg_timer = 0

    def _set_msg(self, text, color=C_INFO, duration=180):
        self.msg_text  = text
        self.msg_color = color
        self.msg_timer = duration

    # ── preset loading ────────────────────────────────────────────────────────

    def _load_preset(self, name):
        data   = PRESETS[name]
        labels = data["nodes"]
        n      = len(labels)

        # Place nodes in a nice arc
        cx, cy = (WIDTH - PANEL_W) // 2, HEIGHT // 2
        radius = min(cx, cy) - NODE_R - 40
        self.nodes, self.edges, self.history = [], [], []
        self.typing_node = None
        self.edge_from   = None

        for i, label in enumerate(labels):
            angle = 2 * math.pi * i / n - math.pi / 2
            x = int(cx + radius * math.cos(angle))
            y = int(cy + radius * math.sin(angle))
            node       = Node(i, x, y)
            node.name  = label
            node.typing = False
            self.nodes.append(node)

        # Convert name-based edges to id-based
        name_to_id = {lbl: i for i, lbl in enumerate(labels)}
        for a, b in data["edges"]:
            self.edges.append((name_to_id[a], name_to_id[b]))

        self._set_msg(f"Skenari '{name}' u ngarkua. Shtyp FILLO LOJËN.", C_OK)
        print(f"[PRESET] loaded '{name}'")

    # ── phase transitions ─────────────────────────────────────────────────────

    def _start_game(self):
        if not self.nodes:
            self._set_msg("Shto te pakten nje nyje!", C_ERR)
            return
        if len(self.edges) == 0:
            self._set_msg("Shto te pakten nje brinje!", C_ERR)
            return
        self._confirm_name()
        if has_cycle(self.edges, len(self.nodes)):
            self._set_msg("Grafi ka cikel -- renditja topologjike eshte e pamundur!", C_ERR)
            return
        self.placed     = set()
        self.order      = []
        self.mistakes   = 0
        self.hint_node  = None
        self.start_ts   = time.time()
        self.elapsed    = 0.0
        self.particles  = []
        self.phase      = "game"
        self._set_msg("Kliko nyjet ne rendin topologjik! Ato te gjelbra jane te disponueshme.", C_INFO)
        print("[GAME START]")

    def _back_to_build(self):
        self.phase = "build"
        self._set_msg("", C_INFO)

    def _restart_game(self):
        self._start_game()

    # ── game action ───────────────────────────────────────────────────────────

    def _place_node(self, nid):
        avail = self._available()

        if nid in self.placed:
            self._set_msg(f"'{self.nodes[nid].display_name()}' eshte vendosur tashme.", C_WARN)
            return

        if nid not in avail:
            # Find which predecessors are blocking
            pred  = self._pred()
            block = [self.nodes[u].display_name()
                     for u in pred[nid] if u not in self.placed]
            msg   = f"'{self.nodes[nid].display_name()}' eshte i bllokuar!"
            if block:
                msg += f"  Duhen: {block}"
            self.mistakes += 1
            self.nodes[nid].flash_frames = 22
            self.nodes[nid].flash_color  = C_ERR
            self._set_msg(msg, C_ERR)
            self.hint_node = None
            print(f"[MISTAKE] tried {nid}  blocking={block}")
            return

        self.placed.add(nid)
        self.order.append(nid)
        self.hint_node = None
        self._set_msg(
            f"'{self.nodes[nid].display_name()}' vendosur!  "
            f"({len(self.order)}/{len(self.nodes)})", C_OK)
        print(f"[PLACE] {nid} '{self.nodes[nid].display_name()}'  order={self.order}")

        if len(self.placed) == len(self.nodes):
            self._win()

    def _win(self):
        self.elapsed   = time.time() - self.start_ts
        self.stars     = 3 if self.mistakes == 0 else (2 if self.mistakes <= 2 else 1)
        self.particles = spawn_confetti(140)
        self.phase     = "won"
        mins = int(self.elapsed) // 60
        secs = int(self.elapsed) % 60
        print(f"[WIN]  time={mins:02d}:{secs:02d}  mistakes={self.mistakes}  stars={self.stars}")

    def _show_hint(self):
        avail = self._available()
        if avail:
            self.hint_node = avail[0]
            self._set_msg(
                f"Sugjerim: provo '{self.nodes[avail[0]].display_name()}'", C_WARN)
        else:
            self._set_msg("Nuk ka sugjerim -- provo nje hap tjeter.", C_DIM)

    # ── hit testing ───────────────────────────────────────────────────────────

    def _node_at(self, pos):
        for n in self.nodes:
            if math.hypot(pos[0] - n.x, pos[1] - n.y) < NODE_R + 6:
                return n
        return None

    # ── click dispatcher ──────────────────────────────────────────────────────

    def handle_click(self, pos):
        # Confirm typing before handling any other click
        if self.typing_node and pos[0] < WIDTH - PANEL_W:
            target = self._node_at(pos)
            if target is None or target is not self.typing_node:
                self._confirm_name()

        # ── PANEL ──
        if pos[0] >= WIDTH - PANEL_W:
            if self.phase == "build":
                if self.btn_mode_node.hit(pos):
                    self._confirm_name()
                    self.mode = "node"
                    self.edge_from = None
                    self._sync_mode_buttons()
                elif self.btn_mode_edge.hit(pos):
                    self._confirm_name()
                    self.mode = "edge"
                    self.edge_from = None
                    self._sync_mode_buttons()
                elif self.btn_start.hit(pos):
                    self._start_game()
                else:
                    for i, btn in enumerate(self.preset_btns):
                        if btn.hit(pos):
                            self._load_preset(list(PRESETS.keys())[i])
                            break
            elif self.phase in ("game", "won"):
                if self.phase == "game" and self.btn_hint.hit(pos):
                    self._show_hint()
                elif self.btn_restart.hit(pos):
                    self._restart_game()
                elif self.btn_back.hit(pos):
                    self._back_to_build()
            return

        # ── CANVAS  —  BUILD PHASE ──
        if self.phase == "build":
            node = self._node_at(pos)

            if self.mode == "node":
                if node is None:
                    # Only add if we're not currently typing
                    if self.typing_node is None:
                        self._add_node(pos)
                elif node is self.typing_node:
                    pass          # click on own naming node: ignore
                else:
                    self._confirm_name()

            elif self.mode == "edge":
                if node is not None:
                    if self.edge_from is None:
                        self.edge_from = node.id
                    else:
                        self._add_edge(self.edge_from, node.id)
                        self.edge_from = None

        # ── CANVAS  —  GAME PHASE ──
        elif self.phase == "game":
            node = self._node_at(pos)
            if node is not None:
                self._place_node(node.id)

    # ── keyboard ──────────────────────────────────────────────────────────────

    def handle_key(self, event):
        if event.key == pygame.K_ESCAPE:
            pygame.quit(); sys.exit()

        # Global shortcuts
        if self.phase == "build":
            if event.key == pygame.K_z and self.typing_node is None:
                self._undo()
            elif event.key == pygame.K_c and self.typing_node is None:
                self._reset_graph()
            elif event.key == pygame.K_RETURN or event.key == pygame.K_KP_ENTER:
                self._confirm_name()

        # Typing mode
        if self.typing_node is not None:
            if event.key in (pygame.K_RETURN, pygame.K_KP_ENTER):
                self._confirm_name()
            elif event.key == pygame.K_BACKSPACE:
                self.typing_node.name = self.typing_node.name[:-1]
            elif event.key == pygame.K_ESCAPE:
                self._confirm_name()
            else:
                ch = event.unicode
                if ch and ch.isprintable():
                    if len(self.typing_node.name) < Node.MAX_CHARS:
                        self.typing_node.name += ch

    # ── draw helpers ──────────────────────────────────────────────────────────

    def _timer_str(self):
        t = (time.time() - self.start_ts) if self.start_ts else 0
        m, s = int(t) // 60, int(t) % 60
        return f"{m:02d}:{s:02d}"

    def _draw_node(self, node, avail_set):
        x, y    = node.x, node.y
        nid     = node.id
        ticks   = pygame.time.get_ticks()

        # Determine fill colour
        if self.phase == "game":
            if nid in self.placed:
                fill = C_PLACED
            elif nid in avail_set:
                fill = C_AVAIL
            else:
                fill = C_LOCKED
        elif self.phase == "won":
            fill = C_PLACED
        elif node.typing:
            fill = C_TYPING
        else:
            fill = C_DEFAULT

        # Pulsing ring for available nodes in game
        if self.phase == "game" and nid in avail_set and nid not in self.placed:
            pulse = int(100 + 80 * math.sin(ticks / 350.0 + nid))
            ring_col = (30, min(255, 120 + pulse // 2), 55)
            pygame.draw.circle(screen, ring_col, (x, y), NODE_R + 7, 3)

        # Hint ring
        if self.phase == "game" and nid == self.hint_node:
            pygame.draw.circle(screen, C_WARN, (x, y), NODE_R + 9, 3)

        # Flash ring on wrong click
        if node.flash_frames > 0:
            alpha = min(255, node.flash_frames * 12)
            fc    = tuple(int(c * alpha / 255) for c in node.flash_color)
            pygame.draw.circle(screen, fc, (x, y), NODE_R + 8, 4)
            node.flash_frames -= 1

        # Main circle
        pygame.draw.circle(screen, fill, (x, y), NODE_R)
        pygame.draw.circle(screen, (fill[0]+30, fill[1]+30, fill[2]+30),
                           (x, y), NODE_R, 2)

        # Name text
        label    = node.display_name()
        font     = node.font_for_name()
        txt_surf = font.render(label, True, C_WHITE)
        screen.blit(txt_surf, txt_surf.get_rect(center=(x, y)))

        # Typing cursor
        if node.typing:
            blink_on = (ticks // 500) % 2 == 0
            if blink_on:
                tw = txt_surf.get_width()
                cx = x + tw // 2 + 3
                pygame.draw.line(screen, C_WHITE,
                                 (cx, y - 10), (cx, y + 10), 2)

        # Sequence badge for placed nodes
        if (self.phase in ("game", "won")) and nid in self.placed:
            idx  = self.order.index(nid) + 1
            bx   = x + NODE_R - 2
            by   = y - NODE_R + 2
            pygame.draw.circle(screen, C_WARN, (bx, by), 11)
            num  = FONT_SM.render(str(idx), True, (20, 20, 20))
            screen.blit(num, num.get_rect(center=(bx, by)))

    def _draw_sequence_bar(self):
        # Bottom bar showing the current order
        bar_h  = 60
        bar_y  = HEIGHT - bar_h
        canvas = WIDTH - PANEL_W
        pygame.draw.rect(screen, SEQ_BG, (0, bar_y, canvas, bar_h))
        pygame.draw.line(screen, UI_LINE, (0, bar_y), (canvas, bar_y), 1)

        if not self.order:
            lbl = FONT_SM.render("Rendi juaj do te shfaqet ketu...", True, C_DIM)
            screen.blit(lbl, (16, bar_y + (bar_h - lbl.get_height()) // 2))
            return

        parts = [self.nodes[nid].display_name() for nid in self.order]
        text  = "  ->  ".join(parts)
        lbl   = FONT_MD.render(text, True, C_OK)
        screen.blit(lbl, (16, bar_y + (bar_h - lbl.get_height()) // 2))

    # ── main draw ─────────────────────────────────────────────────────────────

    def draw(self):
        screen.fill(BG)
        canvas_x = WIDTH - PANEL_W
        avail    = set(self._available()) if self.phase in ("game",) else set()

        # Edges
        for u, v in self.edges:
            color = C_EDGE_DONE if (u in self.placed and v in self.placed) else C_EDGE
            draw_arrow(screen, color, self.nodes[u].pos(), self.nodes[v].pos(), 2)

        # Pending edge preview
        if self.phase == "build" and self.mode == "edge" and self.edge_from is not None:
            mx, my = pygame.mouse.get_pos()
            if mx < canvas_x:
                draw_arrow(screen, C_EDGE_PEND,
                           self.nodes[self.edge_from].pos(), (mx, my), 2)

        # Nodes
        for node in self.nodes:
            self._draw_node(node, avail)

        # Confetti (won phase)
        if self.particles:
            for p in self.particles:
                p.update()
                p.draw(screen)
            self.particles = [p for p in self.particles if p.alive()]

        # Sequence bar (game + won)
        if self.phase in ("game", "won"):
            self._draw_sequence_bar()

        # Status message (bottom-left of canvas, above seq bar)
        if self.msg_text and self.msg_timer > 0:
            seq_offset = 64 if self.phase in ("game", "won") else 8
            lbl = FONT_SM.render(self.msg_text, True, self.msg_color)
            screen.blit(lbl, (12, HEIGHT - seq_offset - lbl.get_height() - 4))
            self.msg_timer -= 1

        # Panel separator
        pygame.draw.line(screen, UI_LINE, (canvas_x, 0), (canvas_x, HEIGHT), 2)

        self._draw_panel(canvas_x)
        pygame.display.flip()

    def _draw_panel(self, px):
        pygame.draw.rect(screen, UI_BG, (px, 0, PANEL_W, HEIGHT))
        tx = px + 16

        def t(s, y, color=C_INFO, font=FONT_MD):
            screen.blit(font.render(s, True, color), (tx, y))

        def hline(y):
            pygame.draw.line(screen, UI_LINE, (px+8, y), (WIDTH-8, y), 1)

        # ── BUILD phase ───────────────────────────────────────────────────────
        if self.phase == "build":
            t("Renditjet Topologjike", 12, (215, 225, 255), FONT_XL)
            t("Nderto Grafin Tend", 38, C_DIM, FONT_SM)
            hline(54)

            self.btn_mode_node.draw(screen)
            self.btn_mode_edge.draw(screen)

            hline(144)
            t("Ngarko Skenar:", 148, C_DIM, FONT_SM)
            for btn in self.preset_btns:
                btn.draw(screen)

            self.btn_start.draw(screen)

            y = self.btn_start.rect.bottom + 18
            hline(y)
            y += 8

            t(f"Nyje: {len(self.nodes)}    Brinje: {len(self.edges)}", y, C_DIM)
            y += 22
            mode_lbl = "Shto Nyje  (klik bosh)" if self.mode == "node" \
                       else "Shto Brinje (klik nyje->nyje)"
            t(f"Modo: {mode_lbl}", y, C_INFO)
            y += 30

            if self.typing_node:
                t(f"Shkruaj emrin e nyjes {self.typing_node.id}", y, C_WARN)
                y += 18
                t("Enter = konfirmo", y, C_DIM, FONT_SM)
            else:
                hline(y)
                y += 8
                for hint in ["Klik hapesire -> nyje e re",
                              "Klik nyje->nyje -> brinje",
                              "Z=Undo   C=Reset   ESC=Dil"]:
                    t(hint, y, C_DIM, FONT_SM); y += 17

        # ── GAME phase ────────────────────────────────────────────────────────
        elif self.phase == "game":
            t("FAZA: LOJE", 12, C_WARN, FONT_XL)
            hline(44)

            self.btn_hint.draw(screen)
            self.btn_restart.draw(screen)
            self.btn_back.draw(screen)

            y = 196
            hline(y); y += 10
            t(f"Koha:     {self._timer_str()}", y, C_INFO); y += 22
            t(f"Gabimet:  {self.mistakes}", y,
              C_ERR if self.mistakes > 0 else C_OK); y += 22
            placed_str = f"{len(self.placed)} / {len(self.nodes)}"
            t(f"Vendosur: {placed_str}", y, C_INFO); y += 28

            hline(y); y += 10
            avail_names = [self.nodes[nid].display_name()
                           for nid in self._available()]
            t("Te disponueshme:", y, C_DIM, FONT_SM); y += 18
            for nm in avail_names:
                t(f"  >> {nm}", y, C_AVAIL, FONT_SM); y += 16

            y = HEIGHT - 108
            hline(y); y += 6
            for hint in ["Nyjet jeshile -> vendos tani",
                          "Nyjet e errëta -> bllokuara",
                          "Shikoni shigjetat per vartesine",
                          "ESC = Dil"]:
                t(hint, y, C_DIM, FONT_XS); y += 16

        # ── WON phase ─────────────────────────────────────────────────────────
        elif self.phase == "won":
            t("PERFUNDUAT!", 12, C_OK, FONT_XL)
            stars_str = "* " * self.stars + "  "
            t(stars_str, 42, C_WARN, FONT_LG)
            hline(70)

            y = 80
            mins = int(self.elapsed) // 60
            secs = int(self.elapsed) % 60
            t(f"Koha:    {mins:02d}:{secs:02d}", y, C_INFO); y += 22
            t(f"Gabimet: {self.mistakes}", y,
              C_ERR if self.mistakes > 0 else C_OK); y += 28

            hline(y); y += 10
            t("Renditja juaj:", y, C_DIM, FONT_SM); y += 18
            for nid in self.order:
                t(f"  {self.order.index(nid)+1}. {self.nodes[nid].display_name()}",
                  y, C_OK, FONT_SM)
                y += 16
                if y > HEIGHT - 130:
                    break

            y = HEIGHT - 108
            hline(y); y += 8
            t("Nje renditje tjeter mund te", y, C_DIM, FONT_SM); y += 16
            t("jete gjithashtu e sakte!", y, C_DIM, FONT_SM)

            self.btn_restart.draw(screen)
            self.btn_back.draw(screen)

    # ── main loop ─────────────────────────────────────────────────────────────

    def run(self):
        clock = pygame.time.Clock()
        print("\n" + "="*60)
        print("  Renditjet Topologjike -- Loje  |  ESC = dil")
        print("="*60)
        while True:
            clock.tick(60)
            for event in pygame.event.get():
                if event.type == pygame.QUIT:
                    pygame.quit(); sys.exit()
                if event.type == pygame.KEYDOWN:
                    self.handle_key(event)
                if event.type == pygame.MOUSEBUTTONDOWN:
                    self.handle_click(event.pos)
            self.draw()


if __name__ == "__main__":
    App().run()
````
