## Kujtues teorik — Pseudokodi i leksionit dhe induksioni

```
InsertionSort(A, n)
1   for i ⟵ 1, 2, …, n−1:
2       aktual = A[i]
3       j = i − 1
4       while j ≥ 0 and A[j] > aktual:
5           A[j+1] = A[j]
6           j = j − 1
7       A[j+1] = aktual
```

**Ideja:** mbaje të renditur nënlistën që rritet në të majtë; çdo element i ri vendoset në pozicionin "korrekt".

### Katër përbërësit e induksionit (slides 40–41)

| Përbërësi | Çfarë bëjmë |
|---|---|
| **Hipoteza e induksionit (HI)** | Formulojmë qartë pohimin në varësi të një variabli $i$ |
| **Rasti bazë** | Tregojmë që HI vlen për vlerën fillestare të $i$ |
| **Hapi induktiv** | *I dobët:* HI për $k$ ⇒ HI për $k+1$. *I fortë:* HI për të gjitha vlerat ≤ $k-1$ ⇒ HI për $k$ |
| **Konkluzioni** | HI vlen për të gjitha vlerat e $i$ |

**HI për Insertion Sort (slide 42):** në fund të iteracionit të $i$-të të ciklit `for`,
nënlista $A[:i+1]$ është e renditur **dhe** përmban të njëjtat $i+1$ elemente që ishin fillimisht në ato pozicione.

**Analiza e rastit më të keq (slides 34–35):** algoritmi duhet të jetë korrekt për **çdo** input,
dhe koha e tij matet me inputin **më të keq** — atë që do të zgjidhte një "kundërshtar".

````python
# ── Mjeti i kontrollit ──────────────────────────────────────────────
# Ekzekutojeni këtë qelizë një herë, në fillim. Pas çdo ushtrimi, qeliza
# "kontrollo(...)" teston funksionin tuaj dhe ju tregon nëse është gati.

def kontrollo(titulli, testet):
    try:
        testet()
    except NotImplementedError:
        print(f"⏳ {titulli}: ende pa plotësuar — zëvendësoni `raise NotImplementedError` me kodin tuaj.")
    except AssertionError as e:
        print(f"✗ {titulli}: {e}")
    except Exception as e:
        print(f"✗ {titulli}: kodi u ndal me gabim  {type(e).__name__}: {e}")
    else:
        print(f"✓ {titulli}: të gjitha testet kaluan!")
````

## Ushtrimi 1 — Nga pseudokodi në Python, rresht për rresht

Implementimi më poshtë ndjek **saktësisht** numërimin e rreshtave të pseudokodit.
Funksioni `gjurmo` printon listën në fund të çdo iteracioni të ciklit `for` —
krahasojeni me shembullin e leksionit `[6, 4, 3, 8, 5]` (slides 25–26).

````python
def insertion_sort(A):
    n = len(A)
    for i in range(1, n):                  # rreshti 1
        aktual = A[i]                      # rreshti 2
        j = i - 1                          # rreshti 3
        while j >= 0 and A[j] > aktual:    # rreshti 4
            A[j + 1] = A[j]                # rreshti 5  (spostim djathtas)
            j = j - 1                      # rreshti 6
        A[j + 1] = aktual                  # rreshti 7
    return A


def gjurmo(A):
    A = list(A)
    print(f"Input:             {A}")
    for i in range(1, len(A)):
        aktual, j = A[i], i - 1
        while j >= 0 and A[j] > aktual:
            A[j + 1] = A[j]
            j -= 1
        A[j + 1] = aktual
        e_renditur = A[:i + 1]
        print(f"pas i={i} (aktual={aktual}):  {A}    ← A[:{i + 1}] = {e_renditur} e renditur")
    return A


gjurmo([6, 4, 3, 8, 5]);
````

**Vëzhgoni:** në rreshtin për `i=3` (`aktual=8`) lista nuk ndryshon — cikli `while` nuk ekzekutohet
asnjë herë, sepse $6 < 8$. Kjo është pikërisht situata e slide 26.

### Po nëse algoritmi ka një gabim të vogël?

Më poshtë është një Insertion Sort me **një karakter të gabuar**: `j > 0` në vend të `j >= 0`.
HI e kap gabimin **në iteracionin e saktë ku ndodh** — jo vetëm në fund.

````python
def insertion_sort_gabim(A):
    for i in range(1, len(A)):
        aktual, j = A[i], i - 1
        while j > 0 and A[j] > aktual:          # ← GABIM: duhet j >= 0
            A[j + 1] = A[j]
            j -= 1
        A[j + 1] = aktual
        if A[:i + 1] != sorted(A[:i + 1]):
            print(f"HI dështon pas iteracionit i={i}: A[:{i + 1}] = {A[:i + 1]}")
            return A
    return A

insertion_sort_gabim([6, 4, 3, 8, 5]);
````

### Grafiku: rasti më i mirë, rasti i rastit, rasti më i keq

Grafiku përdor funksionin e numërimit më poshtë (të pavarur nga zgjidhja juaj), që të shihet
edhe pa plotësuar ushtrimin. Vijat e ndërprera janë formulat teorike.

````python
import plotly.graph_objects as go

def _spostime(A):
    A, s = list(A), 0
    for i in range(1, len(A)):
        aktual, j = A[i], i - 1
        while j >= 0 and A[j] > aktual:
            A[j + 1] = A[j]; s += 1; j -= 1
        A[j + 1] = aktual
    return s

random.seed(1)
ns = list(range(2, 201, 6))
me_i_keq   = [_spostime(range(n, 0, -1)) for n in ns]
i_rastit   = [sum(_spostime(random.sample(range(n), n)) for _ in range(5)) / 5 for n in ns]
me_i_mire  = [_spostime(range(n)) for n in ns]

fig = go.Figure()
fig.add_scatter(x=ns, y=me_i_keq,  mode="markers", name="Rasti më i keq (listë zbritëse)")
fig.add_scatter(x=ns, y=i_rastit,  mode="markers", name="Listë e rastit (mesatare e 5 provave)")
fig.add_scatter(x=ns, y=me_i_mire, mode="markers", name="Rasti më i mirë (listë e renditur)")
fig.add_scatter(x=ns, y=[n * (n - 1) / 2 for n in ns], mode="lines", line=dict(dash="dash"), name="n(n−1)/2")
fig.add_scatter(x=ns, y=[n * (n - 1) / 4 for n in ns], mode="lines", line=dict(dash="dot"),  name="n(n−1)/4")
fig.update_layout(title="Insertion Sort — numri i spostimeve (rreshti 5)",
                  xaxis_title="n (gjatësia e listës)", yaxis_title="spostime", height=460)
fig.show()
````

## Ushtrimi 5 — Selection Sort nën të njëjtën lupë

Selection Sort (laboratori kryesor, Ushtrimi 3) ka një HI tjetër. Pas iteracionit të $i$-të:

1. $A[:i+1]$ është e renditur, **dhe**
2. çdo element i $A[:i+1]$ është ≤ çdo elementi të $A[i+1:]$ (pjesa e majtë mban elementet **më të vegjël**).

**Detyra:** plotësoni `assert`-in në fund të çdo iteracioni.
Pastaj përgjigjuni: *pse pika 2 nuk vlen për Insertion Sort?*

````python
def selection_sort_me_HI(A):
    n = len(A)
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            if A[j] < A[min_idx]:
                min_idx = j
        A[i], A[min_idx] = A[min_idx], A[i]
        # PLOTESO: assert për të dyja pjesët e HI (ndihmë: max(A[:i+1]) <= min(A[i+1:]) kur A[i+1:] s'është bosh)
        raise NotImplementedError
    return A


def _testet():
    for _ in range(300):
        L = random.sample(range(100), random.randint(0, 15))
        assert selection_sort_me_HI(list(L)) == sorted(L), f"rezultat i gabuar për {L}"
    import inspect
    assert "assert" in inspect.getsource(selection_sort_me_HI).split("A[i], A[min_idx]")[1], \
        "shtoni assert-in e HI pas shkëmbimit"

kontrollo("Ushtrimi 5 — HI e Selection Sort", _testet)
````
